## Finite, Countable, and Uncountable Sets

💡 **Intuition.** Two sets have the same "size" if you can pair their elements off perfectly, with none left over. The shock of this section: by that standard $\mathbb{N}$, $\mathbb{Z}$, and even $\mathbb{Q}$ are all the *same* size ("countable"), while $\mathbb{R}$ is strictly bigger. Infinity comes in sizes.

Consider sets $A$ and $B$, whose elements may be any objects. If there is a mapping $x \to f(x)$ for $x \in A, f(x) \in B$, then $f$ is said to be a function or mapping of $A$ into $B$. $A$ is called the domain of $f$, and all values that $f$ can take is called the range of $f$.

If $E \subset A, f(E) = \{ f(x): x\in E \}$. $f(E)$ is called the image of $E$ under $f$. It should be noted that $f(A) \subset B$, but if $f(A) = B$ then we say that $f$ maps $A$ onto $B$. (onto is more specific than into).

If $E \subset B, f^{-1}(E) = \{x: f(x) \in E \}$. $f^{-1}(E)$ is called the inverse image or preimage of $E$ under $f$. For $y \in B, f^{-1}(y)$ is the set of all $x \in A$ such that $f(x) = y$. If for $\forall y \in B, | f^{-1}(y) | \le 1$ is said to be a one-to-one (1:1) mapping of $A$ into $B$. 

If $\exists f$ that is onto and 1:1, then it is said that A and B can be put into a 1:1 correspondence (or "equivalence"). This can be expressed as $A \sim B$

A set $A$ is finite if it has at most $N$ elements. A set $A$ is infinite if it is not finite. A set $A$ is countable if $A \sim \mathbb{N}$. A set $A$ is uncountable if $A$ is not finite and not countable.

The most basic operations of sets are the union ($\cup$) and intersection ($\cap$). For sets $A,B: A \cup B = C$ has elements from $A$ or $B$ or both. $A \cap B = C$ has elements only from both $A$ and $B$. For brevity, we assume that you have general knowledge about interactions of sets and set operators.  

## Metric Spaces

💡 **Intuition.** A metric space is just "a set where we can measure distance," abstracting the ruler away from the number line. The key picture: a set is **open** if every one of its points has *wiggle room* — a whole little ball around it still inside the set. A set is **closed** if it contains all the points its members crowd up against (its limit points). "Closed" is not the opposite of "open"; sets can be both or neither.

### Definitions & Notation

A set $X$ whose elements we'll call points are a metric space if $\forall p,q \in X: \exists d(p,q)$ called the distance function or metric such that:

a) $d(p,p) = 0$

b) $d(p,q) > 0; p\neq q$

c) $d(p,q) = d(q,p)$

d) $\forall r \in X: d(p,q) \le d(p,r) + d(r, q)$

Whenever we say segment $(a,b)$, we mean $x: a < x < b$

Most importantly, let's define keywords.

A neighborhood of $p \in X$ is a set $N_r(p)$ consisting of $ \forall q: d(p,q) < r$, $r$ called the radius of the neighborhood.

A point $p$ is a limit point of $E$ ($p \in E'$) if $\forall r: N_r(p) \backslash \{p\} \cap E \neq \emptyset$. 

A point $p$ is an interior point of $E$ ($p \in E^o$) if $\exists r>0: N_r(p) \subset E$.

A set $E$ is closed if $E' \subseteq E$.

A set $E$ is open if $E \subseteq E^o$.

The closure of a set $E$ is $\bar{E} = E \cup E'$

A set $E$ is perfect if $E$ is closed and $E \subseteq E'$

The complement of $E$ ($E^c$) is the set of all $p \in X, p \notin E$.

### Proof: $\forall r>0: N_r(p)$ is an open set

Let $E = N_r(p)$. Let $q \in E$, then $\exists \Delta_q>0 : d(p,q) = r - \Delta_q < r$. 

Let $s \in N_{\Delta_q}(q)$, then $d(p,s) \le d(p,q) + d(q, s) < r - \Delta_q + \Delta_q = r$.

Therefore, $s \in E$ and $q \in E^o$; $\forall q \in E: \exists r: N_r(q) \subset E$.

```python
# A single, easy-to-use function for visualizing the neighborhood argument.
# Minimal knobs: center p, radius r, optional q (else auto-placed), how far s sits inside N_{Δ_q}(q) (s_frac ∈ (0,1)),
# and an angle for s relative to q to keep the picture clear.
#
# Usage examples (run any of these after defining the function):
#   visualize_open_ball()                                  # defaults
#   visualize_open_ball(r=4.0)                             # larger ball
#   visualize_open_ball(q=(1.0, 1.0))                      # specify q
#   visualize_open_ball(s_frac=0.4, theta_deg=120)         # move s around
#   visualize_open_ball(p=(1,-0.5), r=2.2)                 # shift/resize
#
# The plot labels the sets and shows d(p,q)=r-Δ_q, d(q,s)<Δ_q, and d(p,s)≤d(p,q)+d(q,s)<r.

import numpy as np
import matplotlib.pyplot as plt

def visualize_open_ball(p=(0.0, 0.0), r=3.0, q=None, s_frac=0.7, theta_deg=35.0, rng_seed=0):
    """
    Visualize: if q ∈ N_r(p), then ∃ Δ_q > 0 with N_{Δ_q}(q) ⊂ N_r(p).
    
    Parameters
    ----------
    p : tuple(float, float)
        Center of E = N_r(p).
    r : float
        Radius for E = N_r(p); must be > 0.
    q : tuple(float, float) or None
        A point inside N_r(p). If None, a point is chosen automatically.
    s_frac : float in (0,1)
        Where s sits along a radius of N_{Δ_q}(q): d(q, s) = s_frac * Δ_q.
    theta_deg : float
        Direction (degrees) from q to s within N_{Δ_q}(q).
    rng_seed : int
        Seed used only if q is None.
    """
    p = np.asarray(p, dtype=float)
    if r <= 0:
        raise ValueError("r must be positive.")
    if not (0.0 < s_frac < 1.0):
        raise ValueError("s_frac must be in (0,1).")
    
    # If q not provided, pick a reproducible interior point
    if q is None:
        rng = np.random.default_rng(rng_seed)
        # sample within (0.3 * r, 0.8 * r) to avoid edge cases
        rho = rng.uniform(0.3*r, 0.8*r)
        ang = rng.uniform(0, 2*np.pi)
        q = p + rho * np.array([np.cos(ang), np.sin(ang)])
    else:
        q = np.asarray(q, dtype=float)
    
    d_pq = np.linalg.norm(q - p)
    if not (d_pq < r):
        raise ValueError("q must satisfy ||q - p|| < r. Provide an interior q.")
    
    Delta_q = r - d_pq                     # Δ_q > 0
    theta = np.deg2rad(theta_deg)
    s = q + (s_frac * Delta_q) * np.array([np.cos(theta), np.sin(theta)])
    d_ps = np.linalg.norm(s - p)
    d_qs = np.linalg.norm(s - q)
    
    def plot_circle(center, radius, ax, **kwargs):
        t = np.linspace(0, 2*np.pi, 400)
        xy = center.reshape(2,1) + radius * np.vstack((np.cos(t), np.sin(t)))
        ax.plot(xy[0], xy[1], **kwargs)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # E = N_r(p)
    plot_circle(p, r, ax, linewidth=2)
    ax.scatter([p[0]], [p[1]], s=60, zorder=3)
    ax.text(p[0], p[1] + 0.08, "p", ha="center", va="bottom")
    
    # q and its Δ_q-ball
    plot_circle(q, Delta_q, ax, linestyle="--", linewidth=2)
    ax.scatter([q[0]], [q[1]], s=60, zorder=3)
    ax.text(q[0], q[1] + 0.08, "q", ha="center", va="bottom")
    
    # s inside N_{Δ_q}(q)
    ax.scatter([s[0]], [s[1]], s=60, zorder=3)
    ax.text(s[0], s[1] + 0.08, "s", ha="center", va="bottom")
    
    # Triangle-inequality guide segments
    ax.plot([p[0], q[0]], [p[1], q[1]], linestyle=":")
    ax.plot([q[0], s[0]], [q[1], s[1]], linestyle=":")
    ax.plot([p[0], s[0]], [p[1], s[1]], linestyle=":")
    
    # Annotations
    ax.annotate(
        fr"$d(p,q) = {d_pq:.2f} = r - \Delta_q$",
        xy=((p[0]+q[0])/2, (p[1]+q[1])/2),
        xytext=(p[0] + 0.25*r, p[1] + 1.1*r),
        arrowprops=dict(arrowstyle="->"),
        fontsize=11
    )
    ax.annotate(
        fr"$d(q,s) = {d_qs:.2f} < \Delta_q = {Delta_q:.2f}$",
        xy=((q[0]+s[0])/2, (q[1]+s[1])/2),
        xytext=(q[0] + 0.35*r, q[1] + 0.35*r),
        arrowprops=dict(arrowstyle="->"),
        fontsize=11
    )
    ax.annotate(
        fr"$d(p,s) \leq d(p,q)+d(q,s) < r$" + "\n(by triangle inequality)",
        xy=((p[0]+s[0])/2, (p[1]+s[1])/2),
        xytext=(p[0] - 0.7*r, p[1] - 0.8*r),
        arrowprops=dict(arrowstyle="->"),
        fontsize=11
    )
    
    ax.text(p[0], p[1] + r + 0.15, r"$\partial N_r(p)$", ha="center", va="bottom")
    ax.text(q[0] + 0.12, q[1] - 0.12, r"$N_{\Delta_q}(q)$", ha="left", va="top")
    
    # Formatting
    ax.set_aspect("equal", "box")
    lims = r + 1.2
    ax.set_xlim(p[0]-lims, p[0]+lims)
    ax.set_ylim(p[1]-lims, p[1]+lims)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_title(r"Open-ball containment: $N_{\Delta_q}(q)\subset N_r(p)$ with $q\in N_r(p)$")
    plt.show()
    
    return {
        "p": p, "r": float(r), "q": q, "Delta_q": float(Delta_q),
        "s": s, "d_pq": float(d_pq), "d_qs": float(d_qs), "d_ps": float(d_ps)
    }

# Demo call with defaults for immediate visualization
_ = visualize_open_ball()
```

### Proof: $p \in E' : \forall r, | N_r(p) \cap E | = +\infty$

Suppose that for $p \in E': \exists r>0, M \in \mathbb{N}, |N_r(p) \cap E| = M$

Then, for such $r, \exists \min_{m \in M}{ d(p, q_m)} = r_{\min}$. Then, $(N_{r_{\min}}(p) \backslash {p} )\cap E = \emptyset$, which contradicts that $p \in E'$.

```python
# Retry, fixing mathtext (use \setminus -> \backslash, and escape underscores carefully).

import numpy as np
import matplotlib.pyplot as plt

def visualize_limit_point_contradiction(p=(0.0, 0.0), r=3.0, M=5, points=None, rng_seed=1):
    p = np.asarray(p, dtype=float)
    if r <= 0:
        raise ValueError("r must be positive.")
    if points is None:
        rng = np.random.default_rng(rng_seed)
        pts = []
        while len(pts) < M:
            rho = r * np.sqrt(rng.uniform(0.0, 1.0))
            ang = rng.uniform(0.0, 2*np.pi)
            cand = p + rho * np.array([np.cos(ang), np.sin(ang)])
            if np.linalg.norm(cand - p) > 1e-3:
                pts.append(cand)
        Q = np.array(pts)
    else:
        Q = np.array(points, dtype=float).reshape(-1, 2)
        mask = np.linalg.norm(Q - p, axis=1) < r
        Q = Q[mask]
        if len(Q) == 0:
            raise ValueError("No provided points fall inside N_r(p).")
        M = len(Q)
    
    dists = np.linalg.norm(Q - p, axis=1)
    r_min = float(np.min(dists))
    nearest_idx = np.where(np.isclose(dists, r_min))[0]
    
    def plot_circle(center, radius, ax, **kwargs):
        t = np.linspace(0, 2*np.pi, 400)
        xy = center.reshape(2,1) + radius * np.vstack((np.cos(t), np.sin(t)))
        ax.plot(xy[0], xy[1], **kwargs)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    plot_circle(p, r, ax, linewidth=2)
    ax.scatter([p[0]], [p[1]], s=70, zorder=3)
    ax.text(p[0], p[1] + 0.08, "p", ha="center", va="bottom")
    
    ax.scatter(Q[:,0], Q[:,1], s=60, zorder=3)
    for i, (x, y) in enumerate(Q):
        ax.text(x, y + 0.08, f"$q_{{{i+1}}}$", ha="center", va="bottom", fontsize=10)
        ax.plot([p[0], x], [p[1], y], linestyle=":", linewidth=1)
    
    plot_circle(p, r_min, ax, linestyle="--", linewidth=2)
    
    ax.annotate(
        fr"$M = |N_r(p)\cap E| = {M}$ (finite)",
        xy=(p[0], p[1]), xytext=(p[0] - 0.9*r, p[1] + 1.05*r),
        arrowprops=dict(arrowstyle="->"), fontsize=11
    )
    ax.annotate(
        fr"$r_{{\min}}=\min_m d(p,q_m) = {r_min:.2f}$",
        xy=(p[0] + r_min/np.sqrt(2), p[1] + r_min/np.sqrt(2)),
        xytext=(p[0] + 0.35*r, p[1] + 0.35*r),
        arrowprops=dict(arrowstyle="->"), fontsize=11
    )
    ax.text(
        p[0] - 0.95*r, p[1] - 1.05*r,
        r"$(N_{r_{\min}}(p)\backslash\{p\})\cap E=\varnothing$" + "\n" +
        r"$\Rightarrow$ no $E$-points in every punctured nbhd of $p$" + "\n" +
        r"$\Rightarrow$ contradicts $p\in E'$",
        fontsize=11
    )
    
    ax.text(p[0], p[1] + r + 0.15, r"$\partial N_r(p)$", ha="center", va="bottom")
    ax.text(p[0] + 0.05, p[1] + r_min + 0.1, r"$\partial N_{r_{\min}}(p)$", ha="left", va="bottom")
    ax.set_aspect("equal", "box")
    lims = r + 1.2
    ax.set_xlim(p[0]-lims, p[0]+lims)
    ax.set_ylim(p[1]-lims, p[1]+lims)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_title(r"Finite $|N_r(p)\cap E|$ ⇒ $r_{\min}>0$ ⇒ punctured ball empty near $p$ (contradiction)")
    plt.show()
    
    return {
        "p": p, "r": float(r), "M": int(M),
        "points_inside": Q,
        "r_min": r_min,
        "nearest_indices": nearest_idx.tolist()
    }

# Demo with defaults
_ = visualize_limit_point_contradiction()
```

### Proof: $E$ is open iff $E^c$ is closed.

Let $E^c$ be closed, and $x \in E$, then $x \notin E^c$ and $x \notin (E^c)'$ because $\exists r: N_r(x) \cap E = \emptyset$. Therefore $N_r(x) \subset E$, so $x \in E^o$ and $E$ is open.

Let $E$ be open, and $x \in (E^c)'$, then $\forall r: N_r(x) \cap E^c \neq \emptyset$; since $N_r(x)$ contains at least 1 point from $E^c$, then $x \notin E^o$. Since $E$ is open, then $x \in E^c$, so $E^c$ is closed.

### Proof: $E \subset Y \subset X$. $E$ is open relative to $Y$ iff $\exists G \subset X: E = Y \cap G$ (open).

Definition: Let $E \subset Y \subset X$, where $X$ is a metric space. We say that $E$ is open relative to $Y$ if $\forall p \in E, \exists r>0 : N_r^Y(p) = N_r^X(p) \cap Y \subset E$.

Suppose $E$ is open relative to $Y$, then for each $p \in E, \exists r_p$ such that $d(p,q)<r_p , q \in Y$ imply that $q \in E$.

Let $V_p$ be the set of $q \in X$ such that $d(p,q)<r_p$ and define $G = \cup_{p \in E} V_p$, then $G$ is open and $G \subset X$.

Since $\forall p \in E: p \in V_p$ by construction, $E \subset G \cap Y$. Moreover, by construction, $V_p \cap Y \subset E$, so $G \cap Y \subset E$; hence, $E = Y \cap G$.

If $E = G \cap Y$, then $\forall p \in E: \exists r>0, N_r(p) \subset G$; then $V_p \cap Y \subset E$ so that $E$ is open relative to $Y$.

#### Example: $E = (a, b) \cap \mathbb{Q} \subset \mathbb{Q} \subset \mathbb{R}$

$E$ is not open in $\mathbb{R}$ because all $N_r$ contain irrationals, however, $E$ is open relative to $\mathbb{Q}$.

```python
# Relative openness visualization: E ⊂ Y ⊂ X with E open in the subspace Y.
#
# Minimal API:
#   visualize_relative_openness(a=-1.0, b=1.0, num_balls=4)
#     - Uses Y as a smooth curve embedded in X = ℝ² (the graph y = 0.6 sin x).
#     - E is the portion of Y whose x-coordinate lies in (a, b).
#     - Draws open balls V_p in X around sample points p ∈ E so that
#       (N_r^X(p) ∩ Y) ⊂ E. The union G = ⋃_p V_p is visually indicated.
#     - Second panel illustrates the example E=(a,b)∩ℚ ⊂ ℚ ⊂ ℝ:
#       open in ℚ (relative), not open in ℝ.
#
# Examples:
#   visualize_relative_openness()
#   visualize_relative_openness(a=-0.6, b=1.3, num_balls=5)
#
# Returns key computed values to the caller for inspection. 

import numpy as np
import matplotlib.pyplot as plt

def visualize_relative_openness(a=-1.0, b=1.0, num_balls=4):
    if not (a < b):
        raise ValueError("Require a < b.")
    # Curve Y ⊂ X=ℝ² (param x ↦ (x, 0.6 sin x))
    def Y_of_x(x):
        return np.column_stack([x, 0.6*np.sin(x)])
    
    # Domain for drawing Y
    x_all = np.linspace(-3.0, 3.0, 800)
    Y_all = Y_of_x(x_all)
    
    # E segment along the curve where x ∈ (a, b)
    x_E = np.linspace(a, b, 200)
    Y_E = Y_of_x(x_E)
    p_left = Y_of_x(np.array([a]))[0]
    p_right = Y_of_x(np.array([b]))[0]
    
    # Sample points p_i inside E and define radii r_i so that
    # (N_{r_i}^X(p_i) ∩ Y) ⊂ E. A safe geometric choice is
    # r_i = 0.9 * min(||p_i - p_left||, ||p_i - p_right||).
    xs = np.linspace(a + 0.1*(b-a), b - 0.1*(b-a), num_balls)
    P = Y_of_x(xs)
    dist_left = np.linalg.norm(P - p_left, axis=1)
    dist_right = np.linalg.norm(P - p_right, axis=1)
    radii = 0.9*np.minimum(dist_left, dist_right)
    
    def plot_circle(center, radius, ax, **kwargs):
        t = np.linspace(0, 2*np.pi, 400)
        xy = center.reshape(2,1) + radius * np.vstack((np.cos(t), np.sin(t)))
        ax.plot(xy[0], xy[1], **kwargs)
    
    fig = plt.figure(figsize=(9, 10))
    gs = fig.add_gridspec(2, 1, height_ratios=[2, 1.2])
    
    # ---------- Panel A: Subspace picture (E open in Y) ----------
    ax1 = fig.add_subplot(gs[0, 0])
    
    # X is the whole plane; draw Y (the curve)
    ax1.plot(Y_all[:,0], Y_all[:,1], linewidth=2, label="Y (subset of X)")
    # Highlight E = Y ∩ {(x,y): x∈(a,b)}
    ax1.plot(Y_E[:,0], Y_E[:,1], linewidth=4, label="E = Y ∩ {(a,b)×ℝ}")
    
    # Draw boundary points of E on Y
    ax1.scatter([p_left[0], p_right[0]], [p_left[1], p_right[1]], s=60, zorder=3)
    ax1.text(p_left[0], p_left[1]+0.12, "boundary", ha="center")
    ax1.text(p_right[0], p_right[1]+0.12, "boundary", ha="center")
    
    # Place balls V_p around sample points p ∈ E
    for i, (pi, ri) in enumerate(zip(P, radii), start=1):
        plot_circle(pi, ri, ax1, linestyle="--", linewidth=1.8)
        ax1.scatter([pi[0]], [pi[1]], s=60, zorder=3)
        ax1.text(pi[0], pi[1]-0.15, fr"$p_{i}$", ha="center")
    
    # Annotations explaining G = ⋃ V_p and E = Y ∩ G
    ax1.text(a + 0.05*(b-a), 1.5,
             "Each dashed circle is a ball V_p in X.\n"
             "Its intersection with Y lies inside E.\n"
             r"Let $G=\bigcup_{p\in E} V_p$ (open in $X$) ⇒ $E = Y \cap G$.",
             fontsize=11, va="top")
    
    ax1.set_aspect("equal", "box")
    ax1.set_xlim(-3.2, 3.2)
    ax1.set_ylim(-2.1, 2.1)
    ax1.set_title("Panel A — Subspace topology: $E$ open relative to $Y$ via $E=Y\\cap G$")
    ax1.legend(loc="upper left")
    ax1.set_xlabel("x")
    ax1.set_ylabel("y")
    
    # ---------- Panel B: Example with rationals ----------
    ax2 = fig.add_subplot(gs[1, 0])
    
    # Number line (X = ℝ), Y = ℚ ⊂ ℝ (drawn as many tick-like marks)
    ax2.axhline(0.0, linewidth=1.5)
    ax2.set_yticks([])
    ax2.set_ylim(-0.8, 0.8)
    ax2.set_xlim(a-1.5, b+1.5)
    
    # Open interval (a,b) on the line
    ax2.plot([a, b], [0, 0], linewidth=6, alpha=0.25)  # ambient interval on ℝ
    ax2.text(a, 0.25, "a", ha="center")
    ax2.text(b, 0.25, "b", ha="center")
    ax2.scatter([a, b], [0, 0], s=50)
    
    # Sample "rationals" in (a,b): many small markers (conceptual)
    rq = np.linspace(a+0.05*(b-a), b-0.05*(b-a), 20)
    ax2.scatter(rq, np.zeros_like(rq), marker="x", s=30, label=r"points of $E=(a,b)\cap\mathbb{Q}$")
    
    # Show a small real-ball around a rational point contains irrationals
    x0 = 0.5*(a+b)
    r0 = 0.15*(b-a)
    ax2.plot([x0-r0, x0+r0], [0, 0], linestyle="--", linewidth=2)
    ax2.text(x0, -0.45, r"Any real ball $N_r^\mathbb{R}(x_0)$ hits irrationals", ha="center")
    
    # "Irrational" placeholders inside that ball (conceptual)
    irr = np.array([x0 - 0.4*r0, x0 + 0.35*r0])
    ax2.scatter(irr, [0, 0], marker="^", s=60, label="irrational points (concept)")
    
    ax2.text(x0, 0.45,
             r"$E$ is not open in $\mathbb{R}$ (every $N_r$ contains irrationals),"
             "\n" r"but $E$ is open relative to $\mathbb{Q}$ since subspace balls are $(N_r^\mathbb{R}(x)\cap\mathbb{Q})\subset E$.",
             ha="center", fontsize=11)
    
    ax2.set_title("Panel B — Example: $(a,b)\\cap\\mathbb{Q}$ is open in $\\mathbb{Q}$ but not in $\\mathbb{R}$")
    ax2.legend(loc="upper right")
    ax2.set_xlabel("real line  (X = ℝ ; Y = ℚ)")
    
    plt.tight_layout()
    plt.show()
    
    # Return key values
    return {
        "a": float(a), "b": float(b),
        "P_points": P, "radii": radii,
        "left_boundary": p_left, "right_boundary": p_right
    }

# Demo
_ = visualize_relative_openness()
```

## Compact Sets

💡 **Intuition.** Compactness is the property that lets us upgrade "true near every point" to "true everywhere, with one finite bound." Whenever infinitely many open sets blanket a compact set, finitely many of them already suffice — so arguments that would need infinitely many cases collapse into finitely many. That's why compact sets behave almost like finite sets, and why theorems love them.

### Definitions

An open cover of a set $E \subset X$ is a collection $\{ G_\alpha \}$ such that $ E \subset \cup_\alpha G_\alpha$.

A subset $K \subset X$ is said to be compact if every open cover of $K$ contains a finite subcover; formally: $K \subset \cup_n G_{\alpha_n}$

Compactness is important in analysis, especially w.r.t. continuity (later workshops :D). For $E \subset Y \subset X$, $E$ can be open in $Y$ and not in $X$ (as we've shown), therefore this relatively is dependent on the space in which $E$ is embedded. However, compactness behaves better.

```python
# Visualizing open covers, finite subcovers, and why compactness behaves well under subspaces.
#
# Minimal API:
#   visualize_compactness_open_covers(a=-1.0, b=1.0, num_balls=12, n_intervals=6)
#     - Panel 1 (own figure): X = ℝ², Y a smooth curve embedded in X.
#         K = closed sub-arc of Y with x ∈ [a,b].
#         Build an (initially large) open cover of K using many ambient balls V_i in X.
#         Compute and highlight a finite subcover chosen by a simple greedy set-cover.
#         Annotation emphasizes E=Y∩(⋃V_i) and compactness: every open cover ⇒ finite subcover.
#     - Panel 2 (own figure): Non-compact counterexample on ℝ with E=(0,1) covered by
#         G_n = (1/(n+1), 1-1/(n+1)). Union over all n covers (0,1), but no finite subcover.
#
# Returns a dict with details of the construction.
#
# Examples:
#   visualize_compactness_open_covers()
#   visualize_compactness_open_covers(a=-0.5, b=1.6, num_balls=15, n_intervals=8)

import numpy as np
import matplotlib.pyplot as plt

def visualize_compactness_open_covers(a=-1.0, b=1.0, num_balls=12, n_intervals=6):
    if not (a < b):
        raise ValueError("Require a < b.")
    if num_balls < 3:
        raise ValueError("num_balls should be at least 3.")
    if n_intervals < 3:
        raise ValueError("n_intervals should be at least 3.")
    
    # ----- Geometry of Y in X = ℝ² -----
    def Y_of_x(x):
        return np.column_stack([x, 0.6*np.sin(x)])  # smooth embedded curve
    
    # Discretize the entire curve for context
    x_all = np.linspace(-3.2, 3.2, 900)
    Y_all = Y_of_x(x_all)
    
    # Compact set K: closed arc of Y with x in [a, b]
    xK = np.linspace(a, b, 350)
    K_points = Y_of_x(xK)
    left_bd = Y_of_x(np.array([a]))[0]
    right_bd = Y_of_x(np.array([b]))[0]
    
    # Build a generous open cover of K by ambient balls V_i (centers along K with varying radii)
    xs_centers = np.linspace(a, b, num_balls)
    centers = Y_of_x(xs_centers)
    # Radii: safely cover by using a portion of distance to the nearer endpoint,
    # perturbed a bit so we don't get all equal
    dist_left = np.linalg.norm(centers - left_bd, axis=1)
    dist_right = np.linalg.norm(centers - right_bd, axis=1)
    base = 0.55 * np.minimum(dist_left, dist_right)
    jitter = 0.08 * (np.sin(xs_centers) ** 2)  # small, nonnegative variation
    radii = base + jitter
    
    # Compute a finite subcover via greedy set cover on discretized K
    # Universe: indices of sample points along K
    P = K_points
    m = len(P)
    covered = np.zeros(m, dtype=bool)
    cover_sets = []
    for c, r in zip(centers, radii):
        cover_sets.append(np.linalg.norm(P - c, axis=1) < r)
    
    chosen = []
    while not np.all(covered):
        gains = [np.sum(~covered & S) for S in cover_sets]
        best = int(np.argmax(gains))
        if gains[best] == 0:
            # in the unlikely event of numerical gaps, slightly inflate all radii and retry once
            radii *= 1.02
            cover_sets = [np.linalg.norm(P - c, axis=1) < r for c, r in zip(centers, radii)]
            gains = [np.sum(~covered & S) for S in cover_sets]
            best = int(np.argmax(gains))
            if gains[best] == 0:
                break  # give up; should not happen for our construction
        chosen.append(best)
        covered |= cover_sets[best]
        if len(chosen) > num_balls:  # safety
            break
    
    # ------------- Figure 1: Compact K with open cover and finite subcover -------------
    fig1 = plt.figure(figsize=(8.5, 7.0))
    ax1 = fig1.add_subplot(111)
    
    # Draw ambient space curve Y and highlight K
    ax1.plot(Y_all[:,0], Y_all[:,1], linewidth=2, label="Y (subset of X)")
    ax1.plot(K_points[:,0], K_points[:,1], linewidth=4, label="K ⊂ Y (closed arc)")
    ax1.scatter([left_bd[0], right_bd[0]], [left_bd[1], right_bd[1]], s=55, zorder=3)
    ax1.text(left_bd[0], left_bd[1]+0.12, "a (closed)", ha="center")
    ax1.text(right_bd[0], right_bd[1]+0.12, "b (closed)", ha="center")
    
    # Helper for circles
    def plot_circle(center, radius, ax, **kwargs):
        t = np.linspace(0, 2*np.pi, 400)
        xy = center.reshape(2,1) + radius * np.vstack((np.cos(t), np.sin(t)))
        ax.plot(xy[0], xy[1], **kwargs)
    
    # Draw full open cover (dashed circles)
    for c, r in zip(centers, radii):
        plot_circle(c, r, ax1, linestyle="--", linewidth=1.6)
    
    # Highlight a finite subcover (thicker solid circles)
    for idx in chosen:
        plot_circle(centers[idx], radii[idx], ax1, linewidth=2.6)
    
    ax1.text(a + 0.05*(b-a), 1.55,
             "Dashed circles: an open cover {Gα} of K in X\n"
             "Bold circles: a finite subcover G_{α₁},…,G_{α_n}\n"
             r"Compactness: every open cover of $K$ has a finite subcover."
             "\n(Subspace-friendly: if K is compact in X, then K is compact in Y.)",
             fontsize=11, va="top")
    
    ax1.set_aspect("equal", "box")
    ax1.set_xlim(-3.2, 3.2)
    ax1.set_ylim(-2.0, 2.0)
    ax1.set_title("Compactness on a subspace: open cover ⇒ finite subcover (visual)")
    ax1.set_xlabel("x")
    ax1.set_ylabel("y")
    ax1.legend(loc="upper left")
    plt.show()
    
    # ------------- Figure 2: Non-compact example on ℝ: E=(0,1) -------------
    fig2 = plt.figure(figsize=(8.5, 4.5))
    ax2 = fig2.add_subplot(111)
    
    ax2.axhline(0.0, linewidth=1.5)
    ax2.set_yticks([])
    ax2.set_xlim(-0.2, 1.2)
    ax2.set_ylim(-0.8, 0.8)
    
    # Interval (0,1)
    ax2.plot([0, 1], [0, 0], linewidth=6, alpha=0.25)
    ax2.scatter([0, 1], [0, 0], s=55)
    ax2.text(0, 0.25, "0", ha="center")
    ax2.text(1, 0.25, "1", ha="center")
    
    # Open cover G_n = (1/(n+1), 1-1/(n+1))
    ns = np.arange(1, n_intervals+1)
    for n in ns:
        L = 1.0/(n+1)
        R = 1.0 - 1.0/(n+1)
        ax2.plot([L, R], [0, 0], linestyle="--", linewidth=2)
    
    # Show that any finite selection leaves a gap near 0 or 1
    ax2.text(0.5, -0.5,
             "Open cover of (0,1): G_n = (1/(n+1), 1-1/(n+1))\n"
             "⋃_{n≥1} G_n = (0,1), but no finite subcollection covers points\n"
             "arbitrarily close to 0 or 1 ⇒ (0,1) is not compact in ℝ.",
             ha="center", fontsize=11)
    
    ax2.set_title("Non-compact counterexample on ℝ: an open cover with no finite subcover")
    ax2.set_xlabel("real line")
    plt.show()
    
    return {
        "K_arc_endpoints": (left_bd, right_bd),
        "num_cover_sets": int(num_balls),
        "finite_subcover_indices": [int(i) for i in chosen],
        "interval_cover_n": int(n_intervals)
    }

# Demo call
_ = visualize_compactness_open_covers()
```

### Proof: $K \subset Y \subset X$, then $K$ is compact relative to $X$ iff $K$ is compact relative to $Y$.

#### $X \to Y$

Let $K$ be compact relative to $X$, and let $V_\alpha$ be a collection of sets, open relative to $Y$ such that $K \subset \cup_\alpha V_\alpha$. 

Then, $\forall \alpha, \exists G_\alpha$ open relative to $X$ such that $V_\alpha = Y \cap G_\alpha$. 

Since $K$ is compact relative to $X, K \subset \cup_n G_{\alpha_n}$. 

Moreover, $K \subset Y$, so $K \subset \cup_n V_{\alpha_n}$; $K$ is compact relative to $Y$.

#### $Y \to X$

Let $K$ be compact relative to $Y$, $\{G_\alpha \}$ a collection of open subsets of $X$ covering $K$ with $V_\alpha = Y \cap G_\alpha$. 

Then $\exists \{ \alpha_n \}$ for which $K \subset \cup_n V_{\alpha_n}$. By construction, $V_\alpha \subset G_\alpha$, so $K$ is compact relative to $X$.



### Proof: Compact subsets of metric spaces are closed.

Let $K$ be a compact set of a metric space of $X$, then $K^c$ is an open subset of $X$.

Let $p \in X, p\notin K$. If $q \in K$, let $V_q, W_q$ be neighborhoods of $p,q$ respectively with $r < \frac{1}{2} d(p,q)$. Since $K$ is compact, there are finitely many points $ q_1, q_2, ... q_n$ such that $K = \subset \cup_n W_{q_n} = W$ 

If $V = \cap_n V_{q_n}$, then $V$ is a neighborhood of $p$ that does not intersect $W$, so $V \subset K^c$; $p \in (K^c)^o$, so it is open.

### Proof: Closed subsets of compact sets are compact.

Let $F \subset K \subset X$ for $F$ closed in X, and $K$ compact. 

Let $\{ V_\alpha \}$ be an open cover of $F$. Then, $F^c \cup \{ V_\alpha \}$ is an open cover of K. 

By compactness of $K$, choose finite subcover $ K \subset F^c \cup (\cup_{n} V_{\alpha_n})$

$K \cap F = F \subset  (\cup_{n} V_{\alpha_n})$, hence the original cover still covers $F$ and $F$ is compact.

### Proof: If $\{ K_\alpha \}$ is a collection of compact subsets of $X$ where $\cap_{\alpha_n} K_{\alpha_n} \neq \emptyset$, then $\cap_\alpha K_\alpha \neq \emptyset$.

Fix $K_1$ from the collection, with $G_\alpha = K_\alpha^c$. Assume that $\exists x \in K_1: x\notin \{K_{\alpha_n}\}$, then $\{ G_\alpha \}$ forms an open cover of $K_1$; since $K_1$ is compact, $K_1 \subset \cup_n G_{\alpha_n}$, $K_1 \cap(\cap_n K_{\alpha_n}) = \emptyset$ which contraficts the original statement.

### Proof: If $E$ is an infinite subset of $K$, then $E$ has a limit point in $K$.

Suppose no point of $K$ was a limit point of $E$, then $\forall q \in K, \exists r: N_r(q)$ contains at most one point of $E$. Thus, no finite subcollection can cover $E$ or $K$ since $E \subset K$. This contradicts that assumption that $K$ was compact.

### Theorem: If $E \subset \mathbb{R}^k$ has one property, then it must have the other 2.

a) $E$ is closed and bounded

b) $E$ is compact

c) Every finite subset of $E$ has a limit point in $E$.

## Perfect Sets

### Cantor Set

Construct a set as follows: Let $E_0 = [0,1], E_1 = [0,\frac{1}{3}] \cup [\frac{2}{3}, 1], E_2 = [0,\frac{1}{9}] \cup [\frac{2}{9}, \frac{3}{9}] \cup [\frac{6}{9}, \frac{7}{9}] \cup [\frac{8}{9},1], ...$

Continuing this pattern, $E_0 \supset E_1 \supset ... \supset E_n$. Moreover, there are $2^n$ intervals with length $3^{-n}$. Intuitively, this is nonempty, and furthermore it is a union of closed, finite subcollections so it is also compact. 

```python
# Retry, fixing mathtext (avoid \text{}). Use plain text or simple math only.

import numpy as np
import matplotlib.pyplot as plt

def _cantor_intervals(n):
    intervals = [[(0.0, 1.0)]]
    for _ in range(n):
        prev = intervals[-1]
        nxt = []
        for a, b in prev:
            third = (b - a) / 3.0
            nxt.append((a, a + third))
            nxt.append((b - third, b))
        intervals.append(nxt)
    return intervals

def visualize_cantor(n=5):
    if n < 0:
        raise ValueError("n must be a nonnegative integer.")
    intervals = _cantor_intervals(n)
    
    fig, ax = plt.subplots(figsize=(10, 1.6 + 0.65*n))
    
    y0 = 0.0
    dy = 1.0
    for k, level in enumerate(intervals):
        y = y0 + dy*k
        for a, b in level:
            ax.hlines(y, a, b, linewidth=10)
        count = 2**k
        length_each = 3**(-k)
        total_len = (2/3)**k
        ax.text(-0.08, y, f"E_{k}", va="center", ha="right")
        ax.text(1.02, y, f"{count} intervals × {length_each:.4f} each = {total_len:.4f}", va="center")
    
    ax.set_xlim(-0.12, 1.15)
    ax.set_ylim(-0.8, y0 + dy*len(intervals) + 0.1)
    ax.set_xlabel("x in [0,1]")
    ax.set_yticks([])
    ax.set_title("Middle–third Cantor construction: E0 ⊇ E1 ⊇ ⋯ ⊇ En")
    ax.text(0.02, y0 + dy*(len(intervals)-0.45),
            "Facts:  #(intervals in Ek) = 2^k, length per interval = 3^{-k}\n"
            "total length of Ek = (2/3)^k.  Each Ek is a finite union of closed intervals,\n"
            "so compact. The Cantor set C = ⋂_{k≥0} Ek is closed (intersection of closed sets)\n"
            "and is contained in E0 = [0,1] (compact), hence C is compact.",
            fontsize=10)
    
    plt.tight_layout()
    plt.show()
    
    return {f"E_{k}": level for k, level in enumerate(intervals)}

_ = visualize_cantor(n=6)
```

## Connected Sets

💡 **Intuition.** A set is connected if it can't be split into two camps that don't touch. In $\mathbb{R}$ this has a clean answer: the connected sets are exactly the intervals — sets with "no gaps," containing every point between any two of their members.

### Definitions

Two subsets $A,B \subset X$ are said to be separated if both $A \cap \bar{B} , \bar{A} \cap B =\emptyset$. A set $E$ is connected if $E$ is not a union of 2 nonempty separated sets.

### Proof: $E \subset \mathbb{R}$ is connected iff $\forall x, y \in E, x < z < y$ implies that $z\in E$.

Let $x, y \in E$ and $z \in (x, y), z\notin E$, then $E = A_z \cup B_z$ such that $A_z = E \cap (-\infty, z), B_z = E \cap (z, \infty)$.

Since $x \in A_z, y \in B_z$, $A,B \neq \emptyset$. By construction, $A_z, B_z$ are separated. 

Let $E\subset\mathbb{R}$. Suppose $E=A\cup B$ where $A,B$ are **nonempty, disjoint, and separated in the subspace topology** on $E$; i.e.
$
\overline{A}^{\,E}\cap B=\varnothing \quad\text{and}\quad \overline{B}^{\,E}\cap A=\varnothing .
$
Then there exist $x,y\in E$ and a point $z\in (x,y)$ with $z\notin E$. Equivalently, $E$ has a **gap** point $z$ so that
$
E=(E\cap(-\infty,z))\ \cup\ (E\cap(z,\infty)).
$

Pick $a\in A$ and $b\in B$. If necessary, relabel to ensure $a<b$. Define
$
s=\sup\bigl(A\cap[a,b]\bigr), t=\inf\bigl(B\cap[a,b]\bigr).
$
Then $a\le s\le t\le b$.

- **Case 1: $s<t$.** Choose any $z\in(s,t)$. Then $z\notin A$ (since $s$ is an upper bound of $A\cap[a,b]$) and $z\notin B$ (since $t$ is a lower bound of $B\cap[a,b]$). Hence $z\notin E=A\cup B$. Taking $x\in A\cap[a,s]$ and $y\in B\cap[t,b]$ gives $x<z<y$ with $z\notin E$.

- **Case 2: $s=t$.** Call this common value $z$. There exist sequences $a_n\in A\cap[a,b]$ with $a_n\uparrow z$ and $b_n\in B\cap[a,b]$ with $b_n\downarrow z$. Thus $z\in\overline{A}^{\,\mathbb{R}}\cap \overline{B}^{\,\mathbb{R}}$.  
If $z\in E$, then $z\in\overline{A}^{\,E}\cap \overline{B}^{\,E}$, contradicting that $A$ and $B$ are separated in $E$. Therefore $z\notin E$. Again pick $x\in A\cap[a,z)$ and $y\in B\cap(z,b]$ to get $x<z<y$.

In either case, there exist $x,y\in E$ and $z\in(x,y)$ with $z\notin E$, and consequently
$
E=(E\cap(-\infty,z))\ \cup\ (E\cap(z,\infty)),
$
with the two parts nonempty and separated.

```python
# Re-run the visualizer cell cleanly after reset.

import numpy as np
import matplotlib.pyplot as plt

def _coerce_intervals(spec):
    if spec is None:
        return []
    if isinstance(spec, tuple) and len(spec) == 2:
        return [tuple(map(float, spec))]
    return [tuple(map(float, ab)) for ab in spec]

def _plot_intervals(ax, intervals, y=0.0, height=8, label=None):
    for (a,b) in intervals:
        ax.plot([a,b],[y,y], linewidth=height)
    if label is not None and intervals:
        ax.text(intervals[0][0], y+0.15, label, va="bottom")

def _pick_point_in(intervals, default=None):
    if default is not None:
        return float(default)
    if not intervals:
        return None
    widths = [(b-a) for (a,b) in intervals]
    i = int(np.argmax(widths))
    a,b = intervals[i]
    return (a+b)/2.0

def visualize_separation_on_R(E_left=(0.0, 0.35), E_right=(0.65, 1.0),
                              x=None, y=None, z=None, domain=(0.0, 1.0)):
    L = _coerce_intervals(E_left)
    R = _coerce_intervals(E_right)
    if not L or not R:
        raise ValueError("Provide at least one interval for both E_left and E_right.")
    a_domain, b_domain = map(float, domain)
    
    if z is None:
        r_end = max([b for (a,b) in L])
        l_start = min([a for (a,b) in R])
        z = (r_end + l_start)/2.0 if r_end < l_start else r_end
    x0 = _pick_point_in(L, x)
    y0 = _pick_point_in(R, y)
    
    # Figure A
    figA, axA = plt.subplots(figsize=(10, 3.5))
    axA.axhline(0, linewidth=1.5)
    axA.set_yticks([])
    axA.set_xlim(a_domain-0.05*(b_domain-a_domain), b_domain+0.05*(b_domain-a_domain))
    axA.set_ylim(-0.8, 0.8)
    _plot_intervals(axA, L, y=0, height=10, label="A_z = E ∩ (-∞, z)")
    _plot_intervals(axA, R, y=0, height=10, label="B_z = E ∩ (z, ∞)")
    axA.scatter([x0, y0, z], [0, 0, 0], s=60, zorder=3)
    axA.text(x0, 0.25, "x ∈ E", ha="center")
    axA.text(y0, 0.25, "y ∈ E", ha="center")
    axA.text(z, -0.45, "z ∉ E", ha="center")
    axA.plot([z,z], [-0.18, 0.18], linestyle="--", linewidth=2)
    axA.set_title("Forward: gap at z ⇒ E = A_z ∪ B_z (separated)")
    axA.text((a_domain+b_domain)/2, -0.65,
             "A_z and B_z are nonempty (x∈A_z, y∈B_z) and are separated in the subspace topology.",
             ha="center")
    plt.show()
    
    # Figure B
    minE = min(min(a for a,b in L), min(a for a,b in R))
    maxE = max(max(b for a,b in L), max(b for a,b in R))
    s = max([b for (a,b) in L if b <= maxE])
    t = min([a for (a,b) in R if a >= minE])
    if s < t:
        z_conv = (s + t)/2.0
        z_note = "Pick z in (s,t); then z ∉ E."
    else:
        z_conv = s
        z_note = "Here s = t; z is a limit of A from the right and of B from the left, hence z ∉ E."
    
    figB, axB = plt.subplots(figsize=(10, 3.8))
    axB.axhline(0, linewidth=1.5)
    axB.set_yticks([])
    axB.set_xlim(a_domain-0.05*(b_domain-a_domain), b_domain+0.05*(b_domain-a_domain))
    axB.set_ylim(-0.9, 0.9)
    _plot_intervals(axB, L, y=0, height=10, label="A (from separation)")
    _plot_intervals(axB, R, y=0, height=10, label="B (from separation)")
    axB.scatter([s, t, z_conv], [0, 0, 0], s=60, zorder=3)
    axB.text(s, 0.25, "s = sup A", ha="center")
    axB.text(t, 0.25, "t = inf B", ha="center")
    axB.plot([s, t], [0.05, 0.05], linestyle="--", linewidth=2)
    axB.text(z_conv, -0.55, "z ∉ E", ha="center")
    axB.plot([z_conv, z_conv], [-0.2, 0.2], linestyle="--", linewidth=2)
    axB.set_title("Converse: if E = A ∪ B with A,B nonempty and separated ⇒ there is a gap point z ∉ E")
    axB.text((a_domain+b_domain)/2, -0.75,
             f"Let s = sup A and t = inf B within the span of E. {z_note}",
             ha="center")
    plt.show()
    
    return {
        "A_intervals": L, "B_intervals": R,
        "x": x0, "y": y0, "z_forward": z,
        "s_sup_A": s, "t_inf_B": t, "z_converse": z_conv
    }

_ = visualize_separation_on_R()
```

## Finite, Countable, and Uncountable Sets

💡 **Intuition.** Two sets have the same "size" if you can pair their elements off perfectly, with none left over. The shock of this section: by that standard $\mathbb{N}$, $\mathbb{Z}$, and even $\mathbb{Q}$ are all the *same* size ("countable"), while $\mathbb{R}$ is strictly bigger. Infinity comes in sizes.

Consider sets $A$ and $B$, whose elements may be any objects. If there is a mapping $x \to f(x)$ for $x \in A, f(x) \in B$, then $f$ is said to be a function or mapping of $A$ into $B$. $A$ is called the domain of $f$, and all values that $f$ can take is called the range of $f$.

If $E \subset A, f(E) = \{ f(x): x\in E \}$. $f(E)$ is called the image of $E$ under $f$. It should be noted that $f(A) \subset B$, but if $f(A) = B$ then we say that $f$ maps $A$ onto $B$. (onto is more specific than into).

If $E \subset B, f^{-1}(E) = \{x: f(x) \in E \}$. $f^{-1}(E)$ is called the inverse image or preimage of $E$ under $f$. For $y \in B, f^{-1}(y)$ is the set of all $x \in A$ such that $f(x) = y$. If for $\forall y \in B, | f^{-1}(y) | \le 1$ is said to be a one-to-one (1:1) mapping of $A$ into $B$. 

If $\exists f$ that is onto and 1:1, then it is said that A and B can be put into a 1:1 correspondence (or "equivalence"). This can be expressed as $A \sim B$

A set $A$ is finite if it has at most $N$ elements. A set $A$ is infinite if it is not finite. A set $A$ is countable if $A \sim \mathbb{N}$. A set $A$ is uncountable if $A$ is not finite and not countable.

The most basic operations of sets are the union ($\cup$) and intersection ($\cap$). For sets $A,B: A \cup B = C$ has elements from $A$ or $B$ or both. $A \cap B = C$ has elements only from both $A$ and $B$. For brevity, we assume that you have general knowledge about interactions of sets and set operators.  

## Metric Spaces

💡 **Intuition.** A metric space is just "a set where we can measure distance," abstracting the ruler away from the number line. The key picture: a set is **open** if every one of its points has *wiggle room* — a whole little ball around it still inside the set. A set is **closed** if it contains all the points its members crowd up against (its limit points). "Closed" is not the opposite of "open"; sets can be both or neither.

### Definitions & Notation

A set $X$ whose elements we'll call points are a metric space if $\forall p,q \in X: \exists d(p,q)$ called the distance function or metric such that:

a) $d(p,p) = 0$

b) $d(p,q) > 0; p\neq q$

c) $d(p,q) = d(q,p)$

d) $\forall r \in X: d(p,q) \le d(p,r) + d(r, q)$

Whenever we say segment $(a,b)$, we mean $x: a < x < b$

Most importantly, let's define keywords.

A neighborhood of $p \in X$ is a set $N_r(p)$ consisting of $ \forall q: d(p,q) < r$, $r$ called the radius of the neighborhood.

A point $p$ is a limit point of $E$ ($p \in E'$) if $\forall r: N_r(p) \backslash \{p\} \cap E \neq \emptyset$. 

A point $p$ is an interior point of $E$ ($p \in E^o$) if $\exists r>0: N_r(p) \subset E$.

A set $E$ is closed if $E' \subseteq E$.

A set $E$ is open if $E \subseteq E^o$.

The closure of a set $E$ is $\bar{E} = E \cup E'$

A set $E$ is perfect if $E$ is closed and $E \subseteq E'$

The complement of $E$ ($E^c$) is the set of all $p \in X, p \notin E$.

### Proof: $\forall r>0: N_r(p)$ is an open set

Let $E = N_r(p)$. Let $q \in E$, then $\exists \Delta_q>0 : d(p,q) = r - \Delta_q < r$. 

Let $s \in N_{\Delta_q}(q)$, then $d(p,s) \le d(p,q) + d(q, s) < r - \Delta_q + \Delta_q = r$.

Therefore, $s \in E$ and $q \in E^o$; $\forall q \in E: \exists r: N_r(q) \subset E$.

```python
# A single, easy-to-use function for visualizing the neighborhood argument.
# Minimal knobs: center p, radius r, optional q (else auto-placed), how far s sits inside N_{Δ_q}(q) (s_frac ∈ (0,1)),
# and an angle for s relative to q to keep the picture clear.
#
# Usage examples (run any of these after defining the function):
#   visualize_open_ball()                                  # defaults
#   visualize_open_ball(r=4.0)                             # larger ball
#   visualize_open_ball(q=(1.0, 1.0))                      # specify q
#   visualize_open_ball(s_frac=0.4, theta_deg=120)         # move s around
#   visualize_open_ball(p=(1,-0.5), r=2.2)                 # shift/resize
#
# The plot labels the sets and shows d(p,q)=r-Δ_q, d(q,s)<Δ_q, and d(p,s)≤d(p,q)+d(q,s)<r.

import numpy as np
import matplotlib.pyplot as plt

def visualize_open_ball(p=(0.0, 0.0), r=3.0, q=None, s_frac=0.7, theta_deg=35.0, rng_seed=0):
    """
    Visualize: if q ∈ N_r(p), then ∃ Δ_q > 0 with N_{Δ_q}(q) ⊂ N_r(p).
    
    Parameters
    ----------
    p : tuple(float, float)
        Center of E = N_r(p).
    r : float
        Radius for E = N_r(p); must be > 0.
    q : tuple(float, float) or None
        A point inside N_r(p). If None, a point is chosen automatically.
    s_frac : float in (0,1)
        Where s sits along a radius of N_{Δ_q}(q): d(q, s) = s_frac * Δ_q.
    theta_deg : float
        Direction (degrees) from q to s within N_{Δ_q}(q).
    rng_seed : int
        Seed used only if q is None.
    """
    p = np.asarray(p, dtype=float)
    if r <= 0:
        raise ValueError("r must be positive.")
    if not (0.0 < s_frac < 1.0):
        raise ValueError("s_frac must be in (0,1).")
    
    # If q not provided, pick a reproducible interior point
    if q is None:
        rng = np.random.default_rng(rng_seed)
        # sample within (0.3 * r, 0.8 * r) to avoid edge cases
        rho = rng.uniform(0.3*r, 0.8*r)
        ang = rng.uniform(0, 2*np.pi)
        q = p + rho * np.array([np.cos(ang), np.sin(ang)])
    else:
        q = np.asarray(q, dtype=float)
    
    d_pq = np.linalg.norm(q - p)
    if not (d_pq < r):
        raise ValueError("q must satisfy ||q - p|| < r. Provide an interior q.")
    
    Delta_q = r - d_pq                     # Δ_q > 0
    theta = np.deg2rad(theta_deg)
    s = q + (s_frac * Delta_q) * np.array([np.cos(theta), np.sin(theta)])
    d_ps = np.linalg.norm(s - p)
    d_qs = np.linalg.norm(s - q)
    
    def plot_circle(center, radius, ax, **kwargs):
        t = np.linspace(0, 2*np.pi, 400)
        xy = center.reshape(2,1) + radius * np.vstack((np.cos(t), np.sin(t)))
        ax.plot(xy[0], xy[1], **kwargs)
    
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # E = N_r(p)
    plot_circle(p, r, ax, linewidth=2)
    ax.scatter([p[0]], [p[1]], s=60, zorder=3)
    ax.text(p[0], p[1] + 0.08, "p", ha="center", va="bottom")
    
    # q and its Δ_q-ball
    plot_circle(q, Delta_q, ax, linestyle="--", linewidth=2)
    ax.scatter([q[0]], [q[1]], s=60, zorder=3)
    ax.text(q[0], q[1] + 0.08, "q", ha="center", va="bottom")
    
    # s inside N_{Δ_q}(q)
    ax.scatter([s[0]], [s[1]], s=60, zorder=3)
    ax.text(s[0], s[1] + 0.08, "s", ha="center", va="bottom")
    
    # Triangle-inequality guide segments
    ax.plot([p[0], q[0]], [p[1], q[1]], linestyle=":")
    ax.plot([q[0], s[0]], [q[1], s[1]], linestyle=":")
    ax.plot([p[0], s[0]], [p[1], s[1]], linestyle=":")
    
    # Annotations
    ax.annotate(
        fr"$d(p,q) = {d_pq:.2f} = r - \Delta_q$",
        xy=((p[0]+q[0])/2, (p[1]+q[1])/2),
        xytext=(p[0] + 0.25*r, p[1] + 1.1*r),
        arrowprops=dict(arrowstyle="->"),
        fontsize=11
    )
    ax.annotate(
        fr"$d(q,s) = {d_qs:.2f} < \Delta_q = {Delta_q:.2f}$",
        xy=((q[0]+s[0])/2, (q[1]+s[1])/2),
        xytext=(q[0] + 0.35*r, q[1] + 0.35*r),
        arrowprops=dict(arrowstyle="->"),
        fontsize=11
    )
    ax.annotate(
        fr"$d(p,s) \leq d(p,q)+d(q,s) < r$" + "\n(by triangle inequality)",
        xy=((p[0]+s[0])/2, (p[1]+s[1])/2),
        xytext=(p[0] - 0.7*r, p[1] - 0.8*r),
        arrowprops=dict(arrowstyle="->"),
        fontsize=11
    )
    
    ax.text(p[0], p[1] + r + 0.15, r"$\partial N_r(p)$", ha="center", va="bottom")
    ax.text(q[0] + 0.12, q[1] - 0.12, r"$N_{\Delta_q}(q)$", ha="left", va="top")
    
    # Formatting
    ax.set_aspect("equal", "box")
    lims = r + 1.2
    ax.set_xlim(p[0]-lims, p[0]+lims)
    ax.set_ylim(p[1]-lims, p[1]+lims)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_title(r"Open-ball containment: $N_{\Delta_q}(q)\subset N_r(p)$ with $q\in N_r(p)$")
    plt.show()
    
    return {
        "p": p, "r": float(r), "q": q, "Delta_q": float(Delta_q),
        "s": s, "d_pq": float(d_pq), "d_qs": float(d_qs), "d_ps": float(d_ps)
    }

# Demo call with defaults for immediate visualization
_ = visualize_open_ball()
```

### Proof: $E$ is open iff $E^c$ is closed.

Let $E^c$ be closed, and $x \in E$, then $x \notin E^c$ and $x \notin (E^c)'$ because $\exists r: N_r(x) \cap E = \emptyset$. Therefore $N_r(x) \subset E$, so $x \in E^o$ and $E$ is open.

Let $E$ be open, and $x \in (E^c)'$, then $\forall r: N_r(x) \cap E^c \neq \emptyset$; since $N_r(x)$ contains at least 1 point from $E^c$, then $x \notin E^o$. Since $E$ is open, then $x \in E^c$, so $E^c$ is closed.

### Proof: $E \subset Y \subset X$. $E$ is open relative to $Y$ iff $\exists G \subset X: E = Y \cap G$ (open).

Definition: Let $E \subset Y \subset X$, where $X$ is a metric space. We say that $E$ is open relative to $Y$ if $\forall p \in E, \exists r>0 : N_r^Y(p) = N_r^X(p) \cap Y \subset E$.

Suppose $E$ is open relative to $Y$, then for each $p \in E, \exists r_p$ such that $d(p,q)<r_p , q \in Y$ imply that $q \in E$.

Let $V_p$ be the set of $q \in X$ such that $d(p,q)<r_p$ and define $G = \cup_{p \in E} V_p$, then $G$ is open and $G \subset X$.

Since $\forall p \in E: p \in V_p$ by construction, $E \subset G \cap Y$. Moreover, by construction, $V_p \cap Y \subset E$, so $G \cap Y \subset E$; hence, $E = Y \cap G$.

If $E = G \cap Y$, then $\forall p \in E: \exists r>0, N_r(p) \subset G$; then $V_p \cap Y \subset E$ so that $E$ is open relative to $Y$.

## Compact Sets

💡 **Intuition.** Compactness is the property that lets us upgrade "true near every point" to "true everywhere, with one finite bound." Whenever infinitely many open sets blanket a compact set, finitely many of them already suffice — so arguments that would need infinitely many cases collapse into finitely many. That's why compact sets behave almost like finite sets, and why theorems love them.

### Definitions

An open cover of a set $E \subset X$ is a collection $\{ G_\alpha \}$ such that $ E \subset \cup_\alpha G_\alpha$.

A subset $K \subset X$ is said to be compact if every open cover of $K$ contains a finite subcover; formally: $K \subset \cup_n G_{\alpha_n}$

Compactness is important in analysis, especially w.r.t. continuity (later workshops :D). For $E \subset Y \subset X$, $E$ can be open in $Y$ and not in $X$ (as we've shown), therefore this relatively is dependent on the space in which $E$ is embedded. However, compactness behaves better.

### Proof: $K \subset Y \subset X$, then $K$ is compact relative to $X$ iff $K$ is compact relative to $Y$.

#### $X \to Y$

Let $K$ be compact relative to $X$, and let $V_\alpha$ be a collection of sets, open relative to $Y$ such that $K \subset \cup_\alpha V_\alpha$. 

Then, $\forall \alpha, \exists G_\alpha$ open relative to $X$ such that $V_\alpha = Y \cap G_\alpha$. 

Since $K$ is compact relative to $X, K \subset \cup_n G_{\alpha_n}$. 

Moreover, $K \subset Y$, so $K \subset \cup_n V_{\alpha_n}$; $K$ is compact relative to $Y$.

#### $Y \to X$

Let $K$ be compact relative to $Y$, $\{G_\alpha \}$ a collection of open subsets of $X$ covering $K$ with $V_\alpha = Y \cap G_\alpha$. 

Then $\exists \{ \alpha_n \}$ for which $K \subset \cup_n V_{\alpha_n}$. By construction, $V_\alpha \subset G_\alpha$, so $K$ is compact relative to $X$.



### Proof: Compact subsets of metric spaces are closed.

Let $K$ be a compact set of a metric space of $X$, then $K^c$ is an open subset of $X$.

Let $p \in X, p\notin K$. If $q \in K$, let $V_q, W_q$ be neighborhoods of $p,q$ respectively with $r < \frac{1}{2} d(p,q)$. Since $K$ is compact, there are finitely many points $ q_1, q_2, ... q_n$ such that $K = \subset \cup_n W_{q_n} = W$ 

If $V = \cap_n V_{q_n}$, then $V$ is a neighborhood of $p$ that does not intersect $W$, so $V \subset K^c$; $p \in (K^c)^o$, so it is open.

### Proof: Closed subsets of compact sets are compact.

Let $F \subset K \subset X$ for $F$ closed in X, and $K$ compact. 

Let $\{ V_\alpha \}$ be an open cover of $F$. Then, $F^c \cup \{ V_\alpha \}$ is an open cover of K. 

By compactness of $K$, choose finite subcover $ K \subset F^c \cup (\cup_{n} V_{\alpha_n})$

$K \cap F = F \subset  (\cup_{n} V_{\alpha_n})$, hence the original cover still covers $F$ and $F$ is compact.

### Proof: If $\{ K_\alpha \}$ is a collection of compact subsets of $X$ where $\cap_{\alpha_n} K_{\alpha_n} \neq \emptyset$, then $\cap_\alpha K_\alpha \neq \emptyset$.

Fix $K_1$ from the collection, with $G_\alpha = K_\alpha^c$. Assume that $\exists x \in K_1: x\notin \{K_{\alpha_n}\}$, then $\{ G_\alpha \}$ forms an open cover of $K_1$; since $K_1$ is compact, $K_1 \subset \cup_n G_{\alpha_n}$, $K_1 \cap(\cap_n K_{\alpha_n}) = \emptyset$ which contraficts the original statement.

### Proof: If $E$ is an infinite subset of $K$, then $E$ has a limit point in $K$.

Suppose no point of $K$ was a limit point of $E$, then $\forall q \in K, \exists r: N_r(q)$ contains at most one point of $E$. Thus, no finite subcollection can cover $E$ or $K$ since $E \subset K$. This contradicts that assumption that $K$ was compact.

### Theorem: If $E \subset \mathbb{R}^k$ has one property, then it must have the other 2.

a) $E$ is closed and bounded

b) $E$ is compact

c) Every finite subset of $E$ has a limit point in $E$.

## Connected Sets

💡 **Intuition.** A set is connected if it can't be split into two camps that don't touch. In $\mathbb{R}$ this has a clean answer: the connected sets are exactly the intervals — sets with "no gaps," containing every point between any two of their members.

### Definitions

Two subsets $A,B \subset X$ are said to be separated if both $A \cap \bar{B} , \bar{A} \cap B =\emptyset$. A set $E$ is connected if $E$ is not a union of 2 nonempty separated sets.

### Proof: $E \subset \mathbb{R}$ is connected iff $\forall x, y \in E, x < z < y$ implies that $z\in E$.

Let $x, y \in E$ and $z \in (x, y), z\notin E$, then $E = A_z \cup B_z$ such that $A_z = E \cap (-\infty, z), B_z = E \cap (z, \infty)$.

Since $x \in A_z, y \in B_z$, $A,B \neq \emptyset$. By construction, $A_z, B_z$ are separated. 

Let $E\subset\mathbb{R}$. Suppose $E=A\cup B$ where $A,B$ are **nonempty, disjoint, and separated in the subspace topology** on $E$; i.e.
$
\overline{A}^{\,E}\cap B=\varnothing \quad\text{and}\quad \overline{B}^{\,E}\cap A=\varnothing .
$
Then there exist $x,y\in E$ and a point $z\in (x,y)$ with $z\notin E$. Equivalently, $E$ has a **gap** point $z$ so that
$
E=(E\cap(-\infty,z))\ \cup\ (E\cap(z,\infty)).
$

Pick $a\in A$ and $b\in B$. If necessary, relabel to ensure $a<b$. Define
$
s=\sup\bigl(A\cap[a,b]\bigr), t=\inf\bigl(B\cap[a,b]\bigr).
$
Then $a\le s\le t\le b$.

- **Case 1: $s<t$.** Choose any $z\in(s,t)$. Then $z\notin A$ (since $s$ is an upper bound of $A\cap[a,b]$) and $z\notin B$ (since $t$ is a lower bound of $B\cap[a,b]$). Hence $z\notin E=A\cup B$. Taking $x\in A\cap[a,s]$ and $y\in B\cap[t,b]$ gives $x<z<y$ with $z\notin E$.

- **Case 2: $s=t$.** Call this common value $z$. There exist sequences $a_n\in A\cap[a,b]$ with $a_n\uparrow z$ and $b_n\in B\cap[a,b]$ with $b_n\downarrow z$. Thus $z\in\overline{A}^{\,\mathbb{R}}\cap \overline{B}^{\,\mathbb{R}}$.

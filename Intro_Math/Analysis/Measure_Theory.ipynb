**"Why countable?" deserves five minutes of its own.** Finite closure cannot reach a limit, and analysis is the study of limits — you could not even form $\bigcup_n [0, 1 - 1/n]$. Uncountable closure lets you write any set as the union of its singletons, each null, and Vitali walks straight back in. Every open set, closed set, countable set, $G_\delta$, $F_\sigma$ — everything reachable by countably many ordinary operations — is Borel. You cannot *write down* a non-measurable set; Vitali's requires choice and is inherently non-constructive. Engineers may proceed as though everything they meet is measurable, provided they know why the caveat exists.

**Continuity from below is the session's one real proof; make the technique the lesson.** The move is **disjointification**: $B_n = A_n \setminus A_{n-1}$ converts a nested increasing family into a disjoint one, so countable additivity applies. That trick recurs constantly. And flag the failure of continuity from *above* — $A_n = [n,\infty)$, each of infinite measure, empty intersection — because "finite measure somewhere" is the hypothesis students most often drop.

## Motivation: You Can't Measure Everything

💡 **Intuition.** We'd love a function $m$ assigning every subset of $\mathbb{R}$ a length, with three reasonable demands: intervals get their usual length, shifting a set doesn't change its measure, and the measure of a countable disjoint union is the sum of the pieces. **No such $m$ exists on all subsets** — the axiom of choice builds a set (Vitali's) whose measure can be neither zero nor positive without contradiction. The fix is not to weaken the demands but to *shrink the club of measurable sets* — and the σ-algebra is the membership rule for that club.

**Sketch (Vitali).** On $[0,1)$, call $x \sim y$ when $x - y \in \mathbb{Q}$; pick one representative from each equivalence class to form $V$. The rational translates $\{V + q \pmod 1 : q \in \mathbb{Q} \cap [0,1)\}$ are disjoint and cover $[0,1)$. If $m(V) = 0$, countable additivity gives $m([0,1)) = 0$; if $m(V) > 0$, it gives $m([0,1)) = \infty$. Both contradict $m([0,1)) = 1$. $\blacksquare$

## σ-Algebras

**Definition.** A collection $\mathcal{M}$ of subsets of $X$ is a *σ-algebra* if:

1. $X \in \mathcal{M}$;
2. $A \in \mathcal{M} \implies A^c \in \mathcal{M}$ (closed under complement);
3. $A_1, A_2, \dots \in \mathcal{M} \implies \bigcup_{n=1}^{\infty} A_n \in \mathcal{M}$ (closed under **countable** union).

De Morgan then gives countable intersections for free. The pair $(X, \mathcal{M})$ is a *measurable space*.

💡 **Intuition.** Why exactly *countable* operations? Finite is too weak to reach limits — and limits are the whole point of analysis. Uncountable is too strong — every set would be a union of its points and we'd be back to measuring everything, which Vitali forbids. Countable is the Goldilocks closure: strong enough for $\lim$, $\sup$, $\limsup$ ([Sequences & Series](./Numerical_Sequences_and_Series.ipynb)), weak enough to survive.

**Definition (Borel sets).** The *Borel σ-algebra* $\mathcal{B}(\mathbb{R})$ is the smallest σ-algebra containing all open intervals (it exists: intersect all σ-algebras that contain them). Every open set, closed set, countable set, and virtually every set an engineer will ever write down is Borel.

## Measures

**Definition.** A *measure* on $(X, \mathcal{M})$ is $\mu: \mathcal{M} \to [0, \infty]$ with

1. $\mu(\emptyset) = 0$;
2. **Countable additivity:** for disjoint $A_1, A_2, \dots \in \mathcal{M}$,
$$\mu\Big(\bigsqcup_{n=1}^{\infty} A_n\Big) = \sum_{n=1}^{\infty} \mu(A_n).$$

Immediate consequences: *monotonicity* ($A \subseteq B \Rightarrow \mu(A) \le \mu(B)$, write $B = A \sqcup (B \setminus A)$) and *countable subadditivity*.

### Proof: Continuity from below

If $A_1 \subseteq A_2 \subseteq \cdots$, then $\mu\big(\bigcup_n A_n\big) = \lim_n \mu(A_n)$.

Disjointify: let $B_1 = A_1$ and $B_n = A_n \setminus A_{n-1}$. The $B_n$ are disjoint with $\bigsqcup_{k \le n} B_k = A_n$ and $\bigsqcup_k B_k = \bigcup_n A_n$. Then

$$\mu\Big(\bigcup_n A_n\Big) = \sum_{k=1}^{\infty} \mu(B_k) = \lim_n \sum_{k=1}^{n} \mu(B_k) = \lim_n \mu(A_n). \;\blacksquare$$

*(Continuity from above holds too for $A_1 \supseteq A_2 \supseteq \cdots$, but needs $\mu(A_1) < \infty$ — with $A_n = [n, \infty)$ each has infinite measure yet the intersection is empty.)*

## Lebesgue Measure

**Definition (outer measure).** For any $E \subseteq \mathbb{R}$,

$$\lambda^*(E) = \inf \Big\{ \sum_{n=1}^{\infty} \ell(I_n) \;:\; E \subseteq \bigcup_{n=1}^{\infty} I_n, \; I_n \text{ open intervals} \Big\}$$

— the cheapest way to blanket $E$ with countably many intervals. $\lambda^*$ measures *every* set but is only subadditive. **Carathéodory's criterion** selects the sets on which it behaves: $E$ is *measurable* if it splits every test set cleanly,

$$\lambda^*(A) = \lambda^*(A \cap E) + \lambda^*(A \cap E^c) \quad \forall A \subseteq \mathbb{R}.$$

The measurable sets form a σ-algebra containing $\mathcal{B}(\mathbb{R})$, and on it $\lambda = \lambda^*$ is a genuine measure with $\lambda([a,b]) = b - a$ *(proof: Royden ch. 2 or Folland §1.4 — worth one careful read in your life)*.

### Proof: Countable sets are null

Let $E = \{x_1, x_2, \dots\}$ and $\varepsilon > 0$. Cover $x_n$ by an interval of length $\varepsilon / 2^n$; total length $\sum_n \varepsilon/2^n = \varepsilon$. So $\lambda^*(E) \le \varepsilon$ for every $\varepsilon$, hence $\lambda(E) = 0$. In particular $\lambda(\mathbb{Q}) = 0$: the rationals are *invisible to length*. $\blacksquare$

### The Cantor set: uncountable but null

Recall from [Basic Topology](./Basic_Topology.ipynb) that the Cantor set $C$ is uncountable. Yet after $n$ removal stages, $C$ sits inside $2^n$ intervals of length $3^{-n}$, so $\lambda(C) \le (2/3)^n \to 0$. **Cardinality and measure disagree about what "big" means.**

```python
# Watch the Cantor set's covering length collapse while its point count explodes
lengths = [(2/3)**k for k in range(16)]
plt.figure(figsize=(7, 2.4))
plt.semilogy(lengths, "o-")
plt.title("total length $(2/3)^n$ of stage-$n$ Cantor cover → 0 (yet C is uncountable)")
plt.xlabel("stage n"); plt.grid(True)
plt.tight_layout(); plt.show()
```

**What just happened.** On a log axis the covering length is a straight line falling from 1 at stage 0 to $(2/3)^{15} \approx 0.0023$ at stage 15 — a clean geometric decay to zero with no floor. Since $\lambda(C) \le (2/3)^n$ for **every** $n$, and the only non-negative number below every term of a sequence tending to 0 is 0 itself, $\lambda(C) = 0$ exactly. Not "negligibly small": zero.

**Now hold that beside what the Cantor set contains.** Every point with a base-3 expansion using only the digits 0 and 2 — which maps bijectively onto every binary expansion, hence onto $[0,1]$. The set has the **cardinality of the continuum**. It is as numerous as the entire real line and occupies no length whatever.

**That is the whole point of the demo, and it is worth sitting with.** With $\mathbb{Q}$ a student can rationalise the result away — "countable sets are thin, of course they measure zero." The Cantor set removes that escape. It is uncountable *and* null, so the two notions of size are not merely different scales of the same thing; they disagree about which sets are large. Measure counts length; cardinality counts points; neither implies the other in either direction.

**Note also what the straight line on the log axis is saying about the mechanism.** Each stage keeps $2/3$ of the remaining length, so nothing about the limit depends on removing *middle* thirds — remove middle fifths and you get a different null set with the same structure. Vary the fraction removed at stage $n$ so the total removed is less than 1, and you get a **fat Cantor set**: still nowhere dense, still containing no interval, yet of positive measure. Worth mentioning if the room asks whether "full of holes" and "measure zero" are the same thing. They are not.

**And this is why the coming construction is built on null sets.** "Almost everywhere" — the phrase that lets $L^p$ identify functions differing on a null set, and lets $\mathbf{1}_{\mathbb{Q}}$ integrate to 0 while Riemann chokes on it — is doing real work, and the sets it ignores can be as populous as $\mathbb{R}$.

## The Lebesgue Integral

💡 **Intuition.** Riemann slices the **x-axis** and asks "what is $f$ on this strip?" — helpless when $f$ jitters (e.g. $\mathbf{1}_{\mathbb{Q}}$). Lebesgue slices the **y-axis**: "on how big a set is $f \approx y$?" — and only needs that set to be measurable. Riemann counts a till coin-by-coin as they arrive; Lebesgue sorts the coins into denominations first.

**Construction in three steps.**

1. A *simple function* is a finite sum $\varphi = \sum_k c_k \mathbf{1}_{E_k}$ (measurable $E_k$, $c_k \ge 0$); define $\int \varphi \, d\mu = \sum_k c_k \, \mu(E_k)$.
2. For measurable $f \ge 0$: $\int f \, d\mu = \sup \big\{ \int \varphi \, d\mu : 0 \le \varphi \le f, \varphi \text{ simple} \big\}$.
3. General $f$: split $f = f^+ - f^-$ and set $\int f = \int f^+ - \int f^-$, defined when $\int |f| < \infty$ — then $f \in L^1$.

Key fact making step 2 workable: every measurable $f \ge 0$ is the *increasing pointwise limit* of simple functions (chop the y-axis dyadically).

```python
# Dyadic simple-function approximation of f(x) = x^2 on [0,1]: the y-axis chop
x = np.linspace(0, 1, 1000)
f = x**2
plt.figure(figsize=(7.5, 3))
plt.plot(x, f, "k", linewidth=2, label="$f(x)=x^2$")
for n_levels, alpha in [(4, 0.35), (8, 0.6), (16, 0.9)]:
    phi = np.floor(f * n_levels) / n_levels        # round f DOWN to multiples of 1/n
    plt.step(x, phi, where="post", alpha=alpha, label=f"simple, {n_levels} levels")
plt.legend(); plt.title("Simple functions climbing up to f — the heart of the Lebesgue integral")
plt.tight_layout(); plt.show()
```

**What just happened.** Three staircases climb toward $x^2$, and every one of them stays **underneath** the curve — that is `np.floor` doing its job, rounding $f$ *down* to a multiple of $1/n$. Going from 4 levels to 8 to 16 halves the maximum gap each time, from $1/4$ to $1/8$ to $1/16$: pointwise convergence, and *increasing*, which is exactly what step 2 of the construction demands.

**Read the picture along the right axis.** The steps are horizontal, not vertical — the y-axis was chopped into equal bands, and the x-axis intervals were whatever the preimage happened to be. That is the entire difference from Riemann. Riemann chooses the strips and then asks what $f$ does on each; Lebesgue chooses the *values* and then asks how big the set achieving them is. Here $f$ is smooth so the preimages are tidy intervals and the two approaches agree — but nothing in the Lebesgue side ever assumed that.

**Which is where the generality comes from.** Replace $x^2$ with $\mathbf{1}_{\mathbb{Q}}$ and the picture on the Riemann side collapses: every strip contains both rationals and irrationals, so upper sums are 1 and lower sums are 0, forever, and no refinement helps. On the Lebesgue side nothing changes at all — the function takes two values, the preimages are $\mathbb{Q}$ and its complement, both measurable, and the integral is $1 \cdot \lambda(\mathbb{Q}) + 0 \cdot \lambda(\mathbb{Q}^c) = 0$. **The y-axis chop never needed the preimages to be intervals; it only needed them to be measurable.**

**The dyadic refinement is the general recipe, not a trick for this curve.** For any measurable $f \ge 0$, define $\varphi_n = \min(\lfloor 2^n f\rfloor / 2^n,\, n)$. Halving the mesh each time guarantees $\varphi_n \le \varphi_{n+1}$ — this is why *dyadic* bands rather than arbitrary ones — and the cap at $n$ handles unbounded $f$. So every non-negative measurable function is an increasing limit of simple functions, which is the fact step 2 rests on.

**And the supremum in step 2 is now well-founded rather than mysterious.** $\int f = \sup\{\int \varphi : \varphi \text{ simple}, \varphi \le f\}$ looked like it might be reaching for something unattainable. The staircases show it is attained as a limit: MCT (next cell) says $\int \varphi_n \to \int f$, so the supremum is the limit along *any* increasing simple sequence, and the value does not depend on which one you picked.

## The Convergence Theorems

The payoff for all this machinery: theorems that let you **swap $\lim$ and $\int$** — the move DSP makes every time it differentiates under an integral or sums a Fourier series term by term.

**Monotone Convergence (MCT).** If $0 \le f_1 \le f_2 \le \cdots$ and $f_n \to f$ pointwise, then $\int f_n \, d\mu \to \int f \, d\mu$.

**Fatou's Lemma.** For $f_n \ge 0$: $\int \liminf f_n \, d\mu \le \liminf \int f_n \, d\mu$.

**Dominated Convergence (DCT).** If $f_n \to f$ pointwise **and** $|f_n| \le g$ for some $g \in L^1$, then $\int f_n \, d\mu \to \int f \, d\mu$.

The hypotheses are not decoration — drop the dominator and it fails:

```python
# The escaping bump: f_n = n·1_(0, 1/n] → 0 pointwise, but ∫f_n = 1 forever
x = np.linspace(0, 1, 2000)[1:]
plt.figure(figsize=(7.5, 2.8))
for m in [2, 5, 20]:
    plt.plot(x, np.where(x <= 1/m, m, 0), label=f"$f_{{{m}}}$,  ∫=1")
plt.legend(); plt.ylim(0, 22)
plt.title("f_n → 0 pointwise yet ∫f_n = 1: no g ∈ L¹ dominates these — DCT's hypothesis has teeth")
plt.tight_layout(); plt.show()
```

**What just happened.** Three bumps, each a rectangle of height $m$ and width $1/m$ — so each has area exactly **1**, at $m = 2$, $5$, and $20$ alike. Yet fix any $x > 0$ and the bumps eventually slide left past it: once $1/m < x$, $f_m(x) = 0$ and stays 0. So $f_m \to 0$ **pointwise everywhere on $(0,1]$**, while

$$\int f_m = 1 \quad \text{for every } m, \qquad \int \lim f_m = \int 0 = 0.$$

Limit and integral disagree, permanently, by a factor of $\infty$ in ratio terms. **Pointwise convergence alone is not enough** — mass escapes by getting taller faster than it gets narrower, and pointwise convergence cannot see mass that leaves.

**Check each theorem against this sequence, because that is the lesson.** DCT needs a dominating $g \in L^1$ with $|f_m| \le g$ for all $m$. The smallest candidate is $g(x) = \sup_m f_m(x)$, which behaves like $1/x$ near the origin — and $\int_0^1 dx/x = \infty$. No integrable dominator exists, so DCT does not apply, and its conclusion correspondingly fails. **MCT** does not apply either: the sequence is not increasing (at $x = 0.3$ it goes $2 \to 0 \to 0$). **Fatou** does apply — it needs only $f_m \ge 0$ — and it gives $\int \liminf f_m \le \liminf \int f_m$, i.e. $0 \le 1$. True, and pointedly not an equality. Fatou survives precisely because it never promised one.

**So the hypotheses are not bookkeeping; they are the theorem.** A student who remembers only "you can swap $\lim$ and $\int$" will write a false step sooner or later. The correct memory is: *increasing* buys MCT, *dominated* buys DCT, and non-negativity alone buys you an inequality in one direction only.

**This failure mode has a name and a physical reading.** Concentration: the mass escapes to a point (here $x = 0$). The other classic escape is to infinity — $f_m = \mathbf{1}_{[m, m+1]}$, again $\to 0$ pointwise with $\int f_m = 1$, mass walking off the right-hand end. A dominating $g$ forbids both, which is exactly why the hypothesis is stated as an *integrable envelope* rather than as a bound on each $f_m$ separately. Note that each individual $f_m$ here is perfectly bounded; it is the family that has no common integrable ceiling.

**And it is worth noticing this is the same rectangle as the nascent delta.** $n\mathbf{1}_{(0,1/n]}$ is the standard approximation to $\delta(x)$ used throughout DSP — unit area, shrinking support. Which is precisely why $\delta$ is **not a function**: the pointwise limit of these is the zero function, whose integral is 0, not 1. The distributional limit exists only because distributions are defined by their action on test functions, not pointwise. Every "$\int \delta(x)f(x)\,dx = f(0)$" in a signals course is standing on this distinction.

Students who have used all three without naming them will recognise their own working assumptions, and the abstraction stops looking optional.

**Do not skip past the "almost everywhere" identification.** $L^p$ elements are *equivalence classes*, not functions. Two signals differing at a single sample — or on all of $\mathbb{Q}$ — are the **same element**. The null sets from Session 2 are what make the quotient harmless, and this is where the Cantor demo cashes out. Worth noting the practical consequence: "the value of an $L^2$ signal at $t = 3$" is not well-defined, which is why sampling theorems need continuity hypotheses.

**Young's inequality is where the conjugate exponents stop being arbitrary.** $ab \le a^p/p + b^q/q$ is just concavity of $\log$ applied to the weights $1/p$ and $1/q$ — and those weights must sum to 1 for the argument to be a convex combination. That is the *entire* reason $\frac1p + \frac1q = 1$. Say it explicitly; most students carry the condition as a memorised side-constraint for years.

**The normalisation step in Hölder is the reusable idea.** Divide by the norms so both sides have norm 1, prove the unit case, multiply back. It works because the inequality is homogeneous of degree 1 in each argument, and this "reduce to the unit sphere by scaling" move recurs everywhere in analysis. Ask the room where equality holds — when $|f|^p$ and $|g|^q$ are proportional — and note that $p = q = 2$ recovers Cauchy–Schwarz, which they already know from [Hilbert Spaces](../../Intro_DSP/Hilbert_Spaces.ipynb).

**Name the payoff chain, briefly, without proving the last two links.** Hölder ⟹ Minkowski ⟹ $\|\cdot\|_p$ is a genuine norm ⟹ (Riesz–Fischer) $L^p$ is **complete**. Completeness is the whole point: Cauchy sequences of signals converge *to signals*, so limits of approximations stay in the space. Without it, Fourier series would not be guaranteed to converge to anything you could call a signal. Tie back to the completeness of $\mathbb{R}$ in [Sequences & Series](./Numerical_Sequences_and_Series.ipynb) — same theorem, upgraded from numbers to functions.

**Budget real time for the final counterexample cell, because it corrects a claim students carry in.** "$L^1 \subseteq L^2$" is true for **sequences** ($\ell^p$, counting measure) and false on the line — in both directions. The two failure modes are different in kind: spikes kill $L^2$, tails kill $L^1$. Then close on the one containment that does hold, $L^2[a,b] \subseteq L^1[a,b]$ on finite intervals by Hölder against $g = 1$, since that is the version DSP actually uses for finite-duration signals.

## $L^p$ Spaces

**Definition.** For $1 \le p < \infty$: $\|f\|_p = \big( \int |f|^p \, d\mu \big)^{1/p}$, and $L^p(\mu)$ is the set of measurable $f$ with $\|f\|_p < \infty$ (identifying functions equal *almost everywhere* — i.e. off a null set). $\|f\|_\infty$ is the essential sup.

The signal-processing dictionary: $L^1$ = absolutely integrable (transforms defined pointwise), $L^2$ = **finite energy** (where Fourier theory is a unitary map — Parseval), $L^\infty$ = bounded amplitude.

💡 **Intuition.** Hölder's inequality is a precise version of "you can pay for a product with two different currencies": to bound $\int |fg|$, measure $f$ in the $p$-currency and $g$ in the conjugate $q$-currency, $\frac1p + \frac1q = 1$. The mechanism is Young's inequality — the concavity of the logarithm splitting a product into a weighted sum.

### Proof: Hölder's inequality

Claim: $\|fg\|_1 \le \|f\|_p \, \|g\|_q$ for $\frac1p + \frac1q = 1$, $p, q > 1$.

**Young's inequality.** For $a, b \ge 0$: $ab \le \frac{a^p}{p} + \frac{b^q}{q}$. *Proof:* if $ab=0$ it's trivial; otherwise, since $\log$ is concave, $\log\big(\tfrac{a^p}{p} + \tfrac{b^q}{q}\big) \ge \tfrac1p \log a^p + \tfrac1q \log b^q = \log(ab)$.

**Main step.** The inequality is trivial if $\|f\|_p = 0$ (then $f = 0$ a.e.) or either norm is $\infty$. Otherwise normalize: $F = f / \|f\|_p$, $G = g / \|g\|_q$, so $\|F\|_p = \|G\|_q = 1$. Apply Young pointwise and integrate:

$$\int |FG| \, d\mu \le \frac{1}{p}\int |F|^p \, d\mu + \frac{1}{q}\int |G|^q \, d\mu = \frac1p + \frac1q = 1.$$

Multiply back by $\|f\|_p \|g\|_q$. $\blacksquare$

*(Minkowski's inequality $\|f+g\|_p \le \|f\|_p + \|g\|_p$ — the triangle inequality making $\|\cdot\|_p$ a norm — follows from Hölder; Riesz–Fischer then says $L^p$ is **complete**: Cauchy sequences of signals converge to signals, the [completeness story](./Numerical_Sequences_and_Series.ipynb) upgraded from numbers to functions.)*

### Whose $L^1 \subseteq L^2$? Domain matters

The DSP notebook's claim "$L_1 \subseteq L_2$" is about **sequence spaces** $\ell^p$ (counting measure on $\mathbb{Z}$), and there it's true: if $\sum |a_n| < \infty$ then $|a_n| \le 1$ eventually, so $|a_n|^2 \le |a_n|$ and the comparison test finishes it. On the **line**, the inclusion *fails in both directions*:

```python
x = np.linspace(0.001, 4, 4000)
f1 = np.where(x <= 1, 1/np.sqrt(x), 0)     # in L¹ near 0, NOT in L²
f2 = 1/(1 + x)                              # in L² on [0,∞), NOT in L¹

fig, axes = plt.subplots(1, 2, figsize=(9, 2.6))
axes[0].plot(x, f1); axes[0].set_title(r"$x^{-1/2}\mathbf{1}_{(0,1]}$: $\int f=2$ but $\int f^2=\infty$")
axes[1].plot(x, f2); axes[1].set_title(r"$\frac{1}{1+x}$: $\int f^2<\infty$ but $\int f=\infty$")
for ax in axes: ax.grid(True)
plt.tight_layout(); plt.show()
```

**What just happened.** Two functions, two opposite failures — and neither picture looks pathological:

| function | $\int f$ | $\int f^2$ | verdict |
|---|---|---|---|
| $x^{-1/2}\mathbf{1}_{(0,1]}$ | $\int_0^1 x^{-1/2} = 2$ | $\int_0^1 x^{-1} = \infty$ | $L^1$ but **not** $L^2$ |
| $1/(1+x)$ on $[0,\infty)$ | $\log(1+x) \to \infty$ | $\int_0^\infty (1+x)^{-2} = 1$ | $L^2$ but **not** $L^1$ |

**So on the line neither space contains the other.** Not "the inclusion is subtle" — it fails in both directions, with counterexamples an engineer could plausibly write down. Any argument that quietly assumes finite energy implies absolute integrability, or the reverse, is broken on $\mathbb{R}$.

**The two failures are different in kind, and the plots show which is which.** The left panel blows up at the origin: the **spike** is integrable on its own, but squaring turns $x^{-1/2}$ into $x^{-1}$, and $1/x$ is the exact borderline that diverges. The right panel is bounded and perfectly tame everywhere — its problem is at the far end, where a $1/x$ **tail** decays too slowly to sum, while squaring it to $1/x^2$ decays just fast enough. Squaring punishes height and rewards smallness: it makes spikes worse and tails better. That single sentence predicts both rows.

**Which repairs the claim students carry in from DSP.** "$L^1 \subseteq L^2$" is true — for **sequences**. Under counting measure on $\mathbb{Z}$, $\sum|a_n| < \infty$ forces $|a_n| \le 1$ eventually, so $|a_n|^2 \le |a_n|$ and comparison finishes it. The reason the same argument dies on the line is that a *function* has no such height bound: $x^{-1/2}$ is unbounded near 0, so "$|f| \le 1$ eventually" simply is not available. The inclusion was never about $p$ versus $2$; it was about the underlying **measure**.

**And there is one containment that survives, which is the one DSP uses.** On a finite interval, Hölder with $g = 1$ gives $\|f\|_1 \le \|f\|_2 \sqrt{b-a}$, so $L^2[a,b] \subseteq L^1[a,b]$. Finite measure removes the tail failure mode entirely and leaves only the spike one, which $L^2$ already controls. Every finite-duration signal in a signals course lives here — which is why the false general claim survives so long without anyone noticing.

**A useful habit to leave the room with:** before invoking any $L^p$ inclusion, ask what the measure space is. Finite measure, counting measure, and the whole line give three different answers, and the theorem is about the space, not the exponent.

Moral: spikes ruin $L^2$ faster than $L^1$; slow tails ruin $L^1$ faster than $L^2$. On a **finite** interval, Hölder gives $L^2[a,b] \subseteq L^1[a,b]$ — the direction DSP uses for finite-duration signals.

**Lead with the framing sentence, because it reorganizes everything that follows:** "in Python, NumPy arrays are a library; in MATLAB they're the language." A bare `5` really is a 1×1 matrix — there's no separate scalar type to reason about, which is why operations that look identical to NumPy (`*`, indexing, ranges) quietly mean something different. This isn't trivia; it's the single explanation for most of the porting bugs students will hit.

**Drill the three-row comparison table as the session's real content, not a footnote.** In order of how often they bite:
1. `.*`/`*` is the inverse of NumPy's `*`/`@` — the single most common silent-wrong-answer bug when porting code between the two languages, since `A*B` is *always valid syntax* in both, just computing a different thing.
2. 1-based indexing: `A(1,1)` is the first element, and `A(0,1)` is an error, not "the last row" like Python's negative indexing might suggest.
3. Inclusive ranges: `0:5` includes `5`; the NumPy analogue `range(0,5)` (or `np.arange`) does not. This one is easy to miss because it doesn't error — it just silently runs one iteration too many or too few.

**If a student asks "why does MATLAB do it this way":** the honest answer is history — MATLAB predates NumPy by decades and was built around linear-algebra notation from the start (hence "MATrix LABoratory"), while NumPy retro-fitted array semantics onto general-purpose Python. Neither convention is "wrong"; the skill being taught is *checking which language you're in* before trusting `*`.

## 1. Everything Is a Matrix

💡 **Intuition.** In Python, NumPy arrays are a library; in MATLAB they're the *language*. `5` is a 1×1 matrix. This inversion explains all the syntax: `*` is matrix multiply and you ask for elementwise with `.*` — exactly opposite to NumPy's convention. Most porting bugs between the two languages are this one convention flip.

```matlab
A = [1 2 3; 4 5 6];      % 2x3 matrix — semicolon = new row
x = 0:0.1:1;             % range with step (inclusive!)
t = linspace(0, 1, 500); % 500 points, like np.linspace

size(A)                  % 2 3
A'                       % transpose
A(1, :)                  % first ROW  — indexing is 1-based, round brackets
A(:, 2)                  % second column
A(end, end)              % last element: 'end' is a keyword

B = A .* A;              % elementwise  (.* ./ .^)
C = A * A';              % MATRIX multiply (2x3 * 3x2 = 2x2)
```

The three habits that prevent 90% of Python↔MATLAB confusion:

| | MATLAB | NumPy |
|---|---|---|
| Indexing | `A(1,1)`, 1-based | `A[0,0]`, 0-based |
| Elementwise / matrix `*` | `.*` / `*` | `*` / `@` |
| Range end | `0:5` **includes** 5 | `range(0,5)` excludes 5 |

## 2. Scripts, Functions, Plotting

```matlab
% functions can live at the end of a script file (R2016b+) or in their own file
function y = db(power, ref)
    if nargin < 2, ref = 1.0; end        % default argument, MATLAB-style
    y = 10 * log10(power / ref);
end
```

```matlab
fs = 500;  t = 0:1/fs:1-1/fs;
clean = sin(2*pi*5*t);
noisy = clean + 0.4*randn(size(t));

figure;
plot(t, noisy, 'DisplayName', 'noisy'); hold on;
plot(t, clean, 'LineWidth', 2, 'DisplayName', 'clean');
xlabel('time [s]'); ylabel('amplitude');
title('A 5 Hz sine in noise'); legend; grid on;
```

<details><summary>🎓 <b>Teacher notes — Session 2: Signal Processing in MATLAB</b></summary>

**The `fft` recipe is the session's most important "gotcha" translation — walk it line by line against the Python version students already know.** MATLAB's `fft` returns the full two-sided spectrum by default (same as `np.fft.fft`), but the accompanying recipe folds it into a one-sided spectrum by keeping the first half and doubling all but DC/Nyquist — exactly the transformation students saw in Foundations Session 5 when going from `np.fft.fft` to a physically-interpretable one-sided plot. If this recipe feels unfamiliar, that's the point to loop back to Foundations rather than treat it as new MATLAB syntax.

**The Nyquist-normalization trap in §4 deserves its own pause — it's a real, easy-to-make bug, not a stylistic quirk.** MATLAB's filter-design functions (`fir1`, `butter`, ...) take cutoff frequencies normalized to the Nyquist rate (`Wn = f / (fs/2)`, range 0–1), while `scipy.signal` functions called with the `fs=` keyword take plain Hz directly. Passing a raw Hz value where MATLAB expects a normalized one doesn't error — it silently designs a filter with the wrong cutoff, often by a factor of `fs/2`. The notebook's own advice — "check `freqz` every time" — should be treated as a hard rule, not a suggestion: plot the actual frequency response and confirm the -3 dB point lands where you intended, every single time you design a filter in MATLAB.

**`zplane` (pole-zero plot) is worth connecting explicitly to stability, since it's easy to treat as just another visualization:** poles inside the unit circle mean a stable IIR filter; a pole on or outside the unit circle means the filter's impulse response grows without bound. This is the geometric picture behind why IIR filter design (unlike FIR) can produce an unstable filter if you're not careful with the order or cutoff — FIR filters have no poles (except at the origin) and are therefore unconditionally stable, which is worth stating as the payoff for FIR's higher tap count.

**Close on the meta-lesson from the conclusion:** the DSP *concepts* — spectra, taps, poles, zero-phase filtering — transfer perfectly between MATLAB and Python; only the calling conventions differ. A student who understands `filtfilt` in one language already understands it in the other.
</details>

## 3. The Spectrum

💡 **Intuition.** `fft` is the same radix-2 machine derived in [Foundations Session 5](../../Intro_DSP/Foundations_of_Signal_Processing_1.ipynb) — MATLAB just returns the *two-sided* spectrum, so the standard recipe keeps the first half and doubles it (except DC and Nyquist).

```matlab
X = fft(noisy);
N = numel(noisy);
f = (0:N/2) * fs / N;                    % one-sided frequency axis
P = abs(X(1:N/2+1)) / N;
P(2:end-1) = 2 * P(2:end-1);             % fold negative frequencies in

figure; plot(f, P); xlim([0 50]);
xlabel('frequency [Hz]'); ylabel('|X(f)|');
title('The 5 Hz peak pops right out of the noise');
```

## 4. Filters (Signal Processing Toolbox)

The [Filter Design](../../Intro_DSP/Filter_Design.ipynb) workflow, line for line:

```matlab
% FIR low-pass: 101 taps, 100 Hz cutoff at fs = 1000  (≈ scipy firwin)
fs = 1000;
taps = fir1(100, 100/(fs/2));            % order 100 → 101 taps; freqs normalized to Nyquist!
freqz(taps, 1, 2048, fs);                % magnitude AND phase — read both

% IIR Butterworth, order 4  (≈ scipy butter)
[b, a] = butter(4, 100/(fs/2));
fvtool(b, a);                            % interactive response explorer
zplane(b, a);                            % poles inside the unit circle ⇒ stable

% zero-phase filtering (≈ scipy filtfilt)
y = filtfilt(b, a, noisy);
```

⚠️ The classic trap: MATLAB normalizes frequencies to the **Nyquist** rate (`Wn = f/(fs/2)`),
while `scipy` (with the `fs=` argument) takes plain Hz. Wrong normalization = silently wrong
cutoff — check `freqz` *every time*.

## 2. The Language Core

### 2.1. Variables & Types

Python variables are *labels stuck onto objects* — you don't declare types, the object itself knows what it is. The core types: `int`, `float`, `str`, `bool`.

```python
x = 42          # int
y = 3.14        # float
name = "Fourier"  # str
ready = True    # bool

print(type(x), type(y), type(name), type(ready))
```

Arithmetic works as expected, with two divisions worth memorizing: `/` always gives a float, `//` floors to an integer, and `%` gives the remainder. `**` is exponentiation.

```python
print(7 / 2)    # 3.5   true division
print(7 // 2)   # 3     floor division
print(7 % 2)    # 1     remainder
print(2 ** 10)  # 1024  power
```

**What just happened.** `7 / 2` gave `3.5` (a `float`, always — even `4 / 2` would give `2.0`, not `2`), `7 // 2` floored that same division to `3`, and `7 % 2` gave the leftover `1` (and indeed `3*2 + 1 = 7`, so floor-division and modulo are consistent with each other by construction). `2 ** 10 = 1024` is ordinary exponentiation, with no separate `pow()` call needed. The two operators worth remembering by shape rather than name: `//` and `%` always satisfy `a == (a // b) * b + (a % b)` for any `a`, `b` — useful for sanity-checking either one if you forget which is which mid-derivation.

### 2.2. Strings

Strings are sequences: you can index them, slice them, and glue them together. *f-strings* (the `f"..."` form) are the standard way to format output.

```python
signal = "sinusoid"
print(signal[0], signal[-1])   # first and last character
print(signal[:4])              # slice: characters 0..3
print(f"A {signal} has {len(signal)} letters.")
```

**What just happened.** `signal[0]` and `signal[-1]` returned the first and last characters (`'s'` and `'d'`) — negative indices count backward from the end, so `-1` always means "the last one" regardless of the string's length. `signal[:4]` sliced the first four characters (`'sinu'`); Python slices are half-open, so `[:4]` means "up to but not including index 4," which is why it returns exactly 4 characters, not 5. The f-string interpolated both `signal` and `len(signal)` directly into the output string — `len("sinusoid")` is 8, matching the printed sentence.

### 2.3. Control Flow

Python uses **indentation** instead of braces — the whitespace *is* the syntax. If you took [Intro to C](../Intro_C.ipynb), everything here will feel familiar, just lighter.

```python
for k in range(5):          # k = 0, 1, 2, 3, 4
    if k % 2 == 0:
        print(k, "even")
    else:
        print(k, "odd")
```

```python
# while loops: collatz steps from 27
steps, m = 0, 27
while m != 1:
    m = m // 2 if m % 2 == 0 else 3 * m + 1
    steps += 1
print(f"27 reaches 1 in {steps} steps")
```

**What just happened.** Starting from 27, the Collatz rule (halve if even, `3n+1` if odd) took 111 steps to reach 1 — and along the way it actually climbs *above* 27 several times (27 → 82 → 41 → 124 → ...) before eventually collapsing down, which is part of why the sequence looks unpredictable rather than monotonically shrinking. The `while m != 1` loop is a genuine leap of faith: it terminates for every starting value ever checked by computer (including numbers far larger than 27), but *no one has proven it terminates for all positive integers* — the Collatz conjecture is still open. This cell is a working example of a program whose termination the field cannot currently guarantee in general, even though it visibly halts right here.

### 2.4. Functions

Same idea as in every language: name a computation, give it inputs, return outputs. Default arguments and returning multiple values are everyday Python.

```python
def db(power, reference=1.0):
    """Convert a power ratio to decibels."""
    import math
    return 10 * math.log10(power / reference)

print(db(100))        # 20 dB
print(db(2, 1))       # ~3 dB — the famous "3 dB point" from filter design
```

**What just happened.** `db(100)` used the default `reference=1.0` and returned exactly `20.0` — $10\log_{10}(100/1) = 10 \times 2 = 20$. `db(2, 1)` returned `3.010...`, the famous "3 dB point": a power ratio of 2 is *defined* as +3 dB (to two decimal places), which is why "half power" and "-3 dB" are used interchangeably everywhere in filter design — this function is computing the exact quantity that phrase refers to, not an approximation of it.

💡 **Intuition.** A *comprehension* is a for-loop folded into a single expression: `[f(x) for x in xs if cond(x)]` reads as "the list of f(x), for each x, keeping only those passing the condition." It's the idiom you'll see everywhere in scientific Python.

```python
squares = [k**2 for k in range(10)]
even_squares = [s for s in squares if s % 2 == 0]
print(squares)
print(even_squares)
```

**What just happened.** The first comprehension built all ten perfect squares from 0² through 9² in one expression, no explicit loop variable accumulation needed. The second comprehension then filtered *that* list down to the five even squares (`0, 4, 16, 36, 64`) — notice these are exactly the squares of the even inputs (`0, 2, 4, 6, 8`), since squaring preserves parity. Both comprehensions could equally be written as an explicit `for` loop with `.append()` calls; the comprehension is the same computation, just read left-to-right as "what to build, for what, filtered by what" instead of "start empty, loop, conditionally append."

**The four-container table is the reference students will come back to for the rest of the curriculum — make sure the "why," not just the "what," lands.** The load-bearing distinction is mutability: a `tuple` can be a dict key or set member because it's hashable (immutable); a `list` cannot. This is the same $O(1)$-vs-$O(n)$ distinction the Data Structures in C workshop covers for trees and heaps, applied here to hashing instead.

**Classes: frame `self` as "the object doing the calling," not magic.** `s.duration()` is syntactic sugar for `Signal.duration(s)` — `self` is just the first parameter, filled in automatically. This demystifies both `self` and, later, why forgetting it in a method signature produces a confusing "missing argument" error. ## 3. Data Structures

### 3.1. Lists, Tuples, Dicts, Sets

Four containers cover nearly everything:

| Container | Mutable? | Best for |
|---|---|---|
| `list` | yes | ordered collections you'll grow/change |
| `tuple` | no | fixed groups (coordinates, returns) |
| `dict` | yes | lookup by key |
| `set` | yes | membership tests, de-duplication |

```python
samples = [0.1, 0.5, 0.2, 0.5]        # list
point = (3.0, 4.0)                     # tuple
freqs = {"A4": 440.0, "C5": 523.25}    # dict
unique = set(samples)                  # set: removes the duplicate 0.5

samples.append(0.9)
print(samples)
print(freqs["A4"])
print(unique)
```

**What just happened.** `samples.append(0.9)` mutated the list *in place* — the same `samples` object now has 5 elements, no reassignment needed, which is exactly what "lists are mutable" means concretely. `freqs["A4"]` jumped straight to `440.0` by key, not by scanning. And `set(samples)` — built from the *original* 4-element list before the append — silently dropped one of the two `0.5` entries, since sets only ever store one copy of each distinct value: `{0.1, 0.2, 0.5}` has 3 elements where the source list had 4. That silent de-duplication is the set's entire job, and also the reason to reach for one whenever "how many *distinct* values" is the real question.

💡 **Intuition.** A `dict` is a *phone book*: you don't scan for an entry, you jump straight to it by name. Lookup cost doesn't grow with size — that $O(1)$ behavior comes from hashing the key to a bucket index directly, the same trick behind hash tables in any language.

### 3.2. Classes

A class bundles *data* and the *functions that operate on it*. You've already used them — everything in Python is an object. `__init__` runs at construction; `self` is the object being operated on.

```python
class Signal:
    """A minimal sampled-signal container."""
    def __init__(self, samples, fs):
        self.samples = list(samples)
        self.fs = fs                     # sampling rate in Hz

    def duration(self):
        return len(self.samples) / self.fs

    def scaled(self, gain):
        return Signal([gain * s for s in self.samples], self.fs)

s = Signal([0.0, 0.7, 1.0, 0.7, 0.0], fs=100)
print(f"{s.duration():.2f} s long")
print(s.scaled(2.0).samples)
```

**What just happened.** `s.duration()` computed `len(self.samples) / self.fs = 5 / 100 = 0.05` seconds — the class turned "5 samples at 100 Hz" into the physically meaningful quantity (duration) without the caller needing to remember the formula. `s.scaled(2.0)` didn't modify `s` at all — it built and returned a *brand-new* `Signal` with every sample doubled (`[0.0, 1.4, 2.0, 1.4, 0.0]`), leaving the original untouched. That's a deliberate design choice worth naming: `scaled()` returns a new object rather than mutating `self.samples`, which is why you can chain `s.scaled(2.0).scaled(0.5)` safely without worrying about what state `s` itself ends up in.

## 4. NumPy

💡 **Intuition.** A Python list is a bag of arbitrary objects; a NumPy array is a *contiguous block of numbers of one type* — the same "shelf" picture as C arrays. That layout is what lets NumPy hand whole-array operations to fast vectorized machine code, as demonstrated in [Intro to GPU Systems](../../Intro_GPU/Intro_GPU.ipynb) §2.2.

```python
import numpy as np

t = np.linspace(0, 1, 8)   # 8 points from 0 to 1
x = np.sin(2 * np.pi * t)  # operates on ALL elements at once — no loop
print(t.round(3))
print(x.round(3))
```

### 4.1. Vectorization

The golden rule: **if you're writing a `for` loop over a NumPy array, look for the array operation instead.** It's shorter *and* orders of magnitude faster.

```python
import time

N = 1_000_000
data = np.random.default_rng(42).random(N)

tic = time.perf_counter()
out_loop = [2 * v + 1 for v in data]           # Python loop
t_loop = time.perf_counter() - tic

tic = time.perf_counter()
out_vec = 2 * data + 1                          # vectorized
t_vec = time.perf_counter() - tic

print(f"loop:       {t_loop:.4f} s")
print(f"vectorized: {t_vec:.4f} s  ({t_loop / t_vec:.0f}x faster)")
# exact timings vary by machine; the ratio is the point
```

### 4.2. Shapes, Slicing & Broadcasting

```python
A = np.arange(12).reshape(3, 4)   # 3 rows, 4 columns
print(A)
print(A[1, :])      # row 1
print(A[:, 2])      # column 2
print(A.T.shape)    # transpose: (4, 3)

# broadcasting: subtract each column's mean from the column
print((A - A.mean(axis=0)).round(2))
```

**What just happened.** `A[1, :]` pulled out row 1 (`[4, 5, 6, 7]`) and `A[:, 2]` pulled out column 2 (`[2, 6, 10]`) — same array, two different slicing axes, and `A.T.shape` confirms the transpose swapped the shape from `(3, 4)` to `(4, 3)`. The broadcasting line is the one to slow down on: `A.mean(axis=0)` collapses the *row* axis, producing the length-4 vector of column means `[4, 5, 6, 7]` — and subtracting that length-4 vector from the `(3, 4)` matrix applies it to *every row*, giving `-4` across row 0, `0` across row 1 (which was already exactly the column means), and `+4` across row 2. No loop was written — NumPy stretched the `(4,)` vector against all 3 rows automatically because the trailing dimension matched.

💡 **Intuition.** *Broadcasting* stretches a smaller array across a bigger one without copying: subtracting a length-4 row of means from a (3, 4) matrix applies it to every row. If two shapes match from the right (or one side is 1), NumPy broadcasts.

## 5. Matplotlib

```python
import matplotlib.pyplot as plt

fs = 500                                # sampling rate, Hz
t = np.arange(0, 1, 1 / fs)
clean = np.sin(2 * np.pi * 5 * t)       # 5 Hz sine
noisy = clean + 0.4 * np.random.default_rng(0).standard_normal(t.size)

plt.figure(figsize=(8, 3))
plt.plot(t, noisy, label="noisy", alpha=0.6)
plt.plot(t, clean, label="clean", linewidth=2)
plt.xlabel("time [s]"); plt.ylabel("amplitude")
plt.title("A 5 Hz sine in noise")
plt.legend(); plt.tight_layout(); plt.show()
```

One more — the plot you will make a hundred times in this curriculum: a spectrum. (A preview of [Foundations of Signal Processing](../../Intro_DSP/Foundations_of_Signal_Processing_1.ipynb), where you'll learn what the FFT actually *is*.)

```python
X = np.fft.rfft(noisy)
f = np.fft.rfftfreq(t.size, 1 / fs)

plt.figure(figsize=(8, 3))
plt.plot(f, np.abs(X) / t.size)
plt.xlim(0, 50)
plt.xlabel("frequency [Hz]"); plt.ylabel("|X(f)|")
plt.title("The 5 Hz peak pops right out of the noise")
plt.tight_layout(); plt.show()
```

**What just happened.** `np.fft.rfft` computed the spectrum of the same noisy 5 Hz sine from the plot above, keeping only the non-negative frequencies (the "r" is for "real input" — for real-valued signals the negative-frequency half is a mirror image, so `rfft` skips computing it). `np.fft.rfftfreq` produced the matching frequency axis in Hz, so `plt.plot(f, np.abs(X)/t.size)` shows amplitude vs. real frequency directly. The result is a sharp spike right at 5 Hz rising clearly above a noisy floor — the noise is still there, spread thinly across every frequency, while the actual signal's energy is concentrated at the one frequency it belongs to. This is the visual argument for *why* spectral analysis is useful for noisy signals: a signal buried in time-domain noise can still stand out clearly once you look at the right axis.
